# Why Python?

## 1. What is Python?

Python is a high-level, general-purpose programming language well known for its simplicity and readability.  
Phython was released in 1991 by Guido van Rossum and is widely used in many areas in these days, such as:
- Data analysis
- Machine learning
- Web development
- Automation
- API development
- Scripting
- Scientific computing

### Key Characteristics
1) Simple syntax: Syntax is English-like, making it easy to learn for beginners.  
2) Highly readable: Mandatory indentation ensures that code remains clean and well organized.  
3) Exntensive libraries: It offers a massive ecosystem of other open source libraries, like NumPy, Pandas, Scikit-learn and PyTorch etc.  
4) Interpreted language: Code is executed line-by-line, which makes debugging faster but slower than low-level language like C++.  
---

### Reference: High-Level & Low-Level Programming Languages
#### 1) What Is a High-Level Programming Language?

A **high-level programming language** is designed to be easy for humans to read, write, and understand.

High-level languages hide many complicated details of the computer, such as:

- Memory management
- CPU instructions
- Hardware control
- Low-level system operations

##### Examples of High-Level Languages

- Python
- Java
- JavaScript
- Ruby
- R
- C#
- Swift

#### 2) What Is a Low-Level Programming Language?

A **low-level programming language** is closer to the computer’s hardware.

Low-level languages give the programmer more direct control over memory, CPU instructions, and hardware behavior by using binary, decimal or hexadecimal form. However, they are usually harder to read and write.

##### Examples of Low-Level Languages

- Assembly language
- Machine code

C is often considered closer to low-level than Python or Java, although it is not as low-level as assembly language.

#### 3) Main Difference

| Point | High-Level Language | Low-Level Language |
|---|---|---|
| Human readability | Easier to read | Harder to read |
| Hardware control | Less direct control | More direct control |
| Development speed | Faster | Slower |
| Performance | Usually less optimized | Can be very fast |
| Memory management | Often automatic | Often manual |
| Examples | Python, Java, JavaScript | Assembly, machine code |

#### 4) Simple Analogy

Think of programming like giving instructions to a person.

##### High-Level Language

A high-level instruction might be:

> “Make a sandwich.”

This is easy to understand, but it hides many details.

The person must figure out the steps:

- Take bread
- Add ingredients
- Put the pieces together

##### Low-Level Language

A low-level instruction might be:

> “Move your right hand 20 cm forward, grab one slice of bread, place it on the plate, move your hand back…”

This gives very detailed control, but it takes much longer to write and understand.

#### Summary

A **high-level programming language** is easier for humans to read and write. It hides hardware details and helps programmers build software quickly.

A **low-level programming language** is closer to the hardware. It gives more control and can be faster, but it is harder to write and understand.

In general:

- Use **high-level languages** when productivity, readability, and ease of development are important.
- Use **low-level languages** when speed, memory control, and hardware access are important.

## 2. How Python Differs from Other Languages

Compared with languages such as Java, C++, or C, Python usually requires less code (lines/blocks) to perform the same task.

### Python Example

```python
print("Hello World")
```

### Java Example

```java
public class Main {
    public static void main(String[] args) {
        System.out.println("Hello World");
    }
}
```

Python code is usually concise and easier to understand, especially for beginners.

### Main Differences

| Point | Python | Other Languages |
|---|---|---|
| Syntax | Simple and readable | Often more complex |
| Code length | Usually shorter | Usually longer |
| Speed | Slower than compiled languages | C, C++, Rust, and Java can be faster |
| Development speed | Fast | Usually slower |
| Type system | Dynamically typed | Many languages are statically typed |
| Learning curve | Beginner-friendly | Can be more difficult |
| Main strength | Productivity and flexibility | Performance, strictness, or system-level control |
   
Python is especially useful when development speed, readability, and flexibility are more important than raw execution speed.

## 3. Pros and Cons of Python

### 1) Pros

#### Easy to Learn and Read

Python has simple syntax, so beginners can start coding quickly.

```python
name = "Alice"
print(name)
```

#### Large Ecosystem

Python has many powerful libraries and frameworks.

| Area | Libraries / Frameworks |
|---|---|
| Data analysis | pandas, NumPy |
| Machine learning | scikit-learn, PyTorch, TensorFlow |
| Visualization | matplotlib, plotly, seaborn |
| Web development | Django, Flask, FastAPI |
| Automation | os, pathlib, subprocess |
| Testing | pytest, unittest |

Because of this ecosystem, Python is very productive for real-world projects.

#### Strong for Data and AI

Python is one of the most popular languages for data science, machine learning, and AI.

It is commonly used to:

- Clean data
- Analyze data
- Build machine learning models
- Train deep learning models
- Create dashboards
- Automate data pipelines

#### Cross-Platform

Python runs on many operating systems, including:

- Windows
- macOS
- Linux

This makes it easy to develop and run Python programs in different environments.

#### Good for Automation

Python is very useful for automating repetitive tasks, such as:

- Reading files
- Processing Excel or CSV data
- Calling APIs
- Downloading data
- Sending reports
- Running scheduled jobs

### 2) Cons

#### Slower Execution Speed

Python is usually slower than compiled languages such as C, C++, or Rust.

This is because Python is generally interpreted and focuses more on developer productivity than raw performance.

However, for many tasks, this is not a major problem because Python can use optimized libraries written in C or C++.

Examples:

- NumPy
- pandas
- PyTorch
- TensorFlow

#### Not Ideal for Mobile App Development

Python is not commonly used for building native mobile apps.

For mobile development, other languages are more common.

| Platform | Common Languages |
|---|---|
| iOS | Swift |
| Android | Kotlin, Java |
| Cross-platform | Dart, JavaScript, TypeScript |

#### Memory Usage Can Be High

Python may use more memory than lower-level languages.

This can matter when building systems that need very high performance or run on limited hardware.

#### Runtime Errors Are Easier to Miss

Python is dynamically typed, so some errors may only appear when the program runs.

```python
x = "10"
y = 5

print(x + y)  # Error
```

In statically typed languages, this kind of error may be caught earlier during compilation.

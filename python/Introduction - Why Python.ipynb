# 1. What Is Python?

Python is a high-level, general-purpose programming language known for simple syntax and strong readability.

It was released in 1991 by Guido van Rossum and is now widely used for:

- Data analysis
- Machine learning and AI
- Web development
- Automation
- API development
- Scripting
- Scientific computing

⚠️ <strong>Trainer note:</strong> For beginners, the most important idea is not that Python is "easy". It is that Python lets you express useful ideas with less ceremony.


````python
# Your first Python instruction
print("Hello, World!")

# Press shift+enter to run the code, then you will see "Hello, World!" printed below.
````

## 🗓️ Key Characteristics

| Characteristic | What it means for trainees |
|---|---|
| Simple syntax | Python reads closer to English than many languages |
| Readable structure | Indentation helps code stay organized |
| Extensive libraries | You can use powerful tools like NumPy, pandas, scikit-learn, and PyTorch. It means you don't need to develop functions by yourself from scratch |
| Interpreted execution | Python can run code line by line, which helps with learning and debugging |


⚠️ <strong>Small correction:</strong> Python is productive, but it is not always the fastest language for raw execution speed. Many Python libraries solve this by using optimized C, C++, Rust, or GPU code underneath.

## 🗓️ Reference: High-Level vs Low-Level Languages

A high-level programming language is designed to be easier for humans to read, write, and understand. It hides many computer details such as memory management, CPU instructions, and hardware control.

A low-level language is closer to hardware. It gives more direct control, but it is usually harder to read and write.

| Point | High-Level Language | Low-Level Language |
|---|---|---|
| Human readability | Easier to read | Harder to read |
| Hardware control | Less direct control | More direct control |
| Development speed | Faster | Slower |
| Performance control | Less direct | More direct |
| Memory management | Often automatic | Often manual |
| Examples | Python, Java, JavaScript, R, C# | Assembly, machine code |


⚠️ <strong>Analogy:</strong>  
A high-level instruction is like saying "make a sandwich."  
A low-level instruction describes every tiny movement required to make it happen.

Python usually requires fewer lines of code than languages such as Java, C, or C++ for the same beginner-level task.

## 🗓️ Main Differences

| Point | Python | Other Languages |
|---|---|---|
| Syntax | Simple and readable | Often more formal or verbose |
| Code length | Usually shorter | Usually longer |
| Speed | Slower than compiled languages | C, C++, Rust, and Java can be faster |
| Type system | Dynamically typed | Many are statically typed |
| Learning curve | Beginner-friendly | Relatively slow |
| Main strength | Productivity and flexibility | Performance, strictness, or system-level control |

# 5. Python Installation

There are several ways to install Python.   
The best choice depends on your goal, operating system, and comfort with the terminal.

| Option | Best for | Notes | Reference |
|---|---|---|---|
| Official Python installer | General beginners | Simple and standard | https://www.python.org/downloads/ |
| Miniconda | Learners who want environment management | Smaller Conda installer | https://www.anaconda.com/docs/getting-started/miniconda/install/overview |
| Miniforge | Scientific/open-source package workflows | Uses conda-forge by default | https://github.com/conda-forge/miniforge |
| Anaconda | Beginners who want GUI tools and many packages | Convenient but large | https://www.anaconda.com/download |
| uv | Project-based modern Python development | Fast, uses `.venv` and `pyproject.toml` | https://docs.astral.sh/uv/getting-started/installation/ |
| Package manager | Terminal users | Homebrew, apt, dnf | OS-specific |


⚠️ <strong>Trainer notes:</strong>

For beginners, Anaconda would be a good option; It provides GUI and several tools in a package.  
Most of python libraries for data analysis are included.  
For intermeditate level, Miniconda, Miniforge, uv or other dependency manager like poetry is recommended.

## Option 1. Official Python Installer

### Windows

1. Download Python from https://www.python.org/downloads/.
2. Run the Windows installer.
3. ⚠️ Important: check `Add Python to PATH`. (It may be the end of installation)
4. Click `Install Now`.
5. Open Command Prompt or PowerShell and check:

```bash
python --version
py --version
```

### macOS

1. Download the macOS installer from https://www.python.org/downloads/.
2. Open the `.pkg` installer and follow the steps.
3. Open Terminal and check:

```bash
python3 --version
```

Optional alias if you want to type "python" instead of "python3":

```bash
echo "alias python='python3'" >> ~/.zshrc
```

### Linux

Many Linux systems already include Python. Check first:

```bash
python3 --version
```

Ubuntu / Debian:

```bash
sudo apt update
sudo apt install python3 python3-pip
```

Fedora:

```bash
sudo dnf install python3 python3-pip
```


---

## Option 2. Conda: Miniconda, Miniforge, or Anaconda

Conda is both a package manager and an environment manager.

It can:
- Install Python
- Install libraries
- Create separate environments for different projects
- Manage different Python versions


| Choice | Description | Recommendation |
|---|---|---|
| Miniconda | Minimal Conda installer | Good default choice |
| Miniforge | Minimal installer using conda-forge | Strong choice for scientific packages |
| Anaconda Distribution | Large installer with many tools | Easy but heavy |

### Typical Conda installation check

After installing your chosen Conda distribution, restart the terminal and run:

```bash
conda --version
python --version
```

If you installed Miniforge, you may also check:

```bash
mamba --version
```


## 🗓️ Example of Conda Workflow

Do not do project work directly in the `base` environment; For training purpose, you can use the `base` environment.  
We recommand to create a separate environment for each project.  
Here are steps to create new environment

⚠️ <b> Please replace (project_name) & {python_version} in following intructions. </b>

### Step 1: Create a New Environment

```bash
conda create -n {project_name} python={python_version}
```

### Step 2: Activate the Environment

```bash
conda activate {project_name}
```

Your terminal may show:

```text
{project_name}
```

### Step 3: Check Python

```bash
python --version
```

### Step 4: Install Packages

```bash
conda install pandas numpy matplotlib
```

With Miniforge, you can also use:

```bash
mamba install pandas numpy matplotlib
```

### Step 5: Deactivate

```bash
conda deactivate
```

# 6. How To Execute Python

After Python is installed, you can run Python in several ways.

| Method | Best for | What trainees should remember |
|---|---|---|
| Interactive mode | Testing one or two lines quickly | Great for experiments |
| Python file `.py` | Reusable programs | Best for scripts and apps |
| VS Code | Writing projects | Editor plus terminal workflow |
| Jupyter Notebook | Learning, data analysis, visualization | Code, notes, and output together |
| VS Code Notebook | Notebook workflow inside VS Code | Useful for class and project work |


## Method 1. Interactive Mode

Interactive mode lets you type Python code one line at a time.  
Type below on your terminal to start the interactive mode.  

Windows:

```bash
python
py
```

macOS / Linux:

```bash
python3
```

Conda environment:

```bash
conda activate py-practice
python
```

uv project:

```bash
uv run python
```

You can type python scripts in the terminal and it will return results.

Example: Type following script and press enter.

```python
print("Hello, Python!")
```

If you want to exit the interactive mode, then type below:

```python
exit()
```


---

## Method 2. Run A Python File

Create a file named `hello.py` by any text editors and type following script:

```python
print("Hello, Python!")
```

Run it with commands below in your terminal:

<b> Windows: </b>

```bash
python hello.py
py hello.py
```

<b> macOS / Linux: </b>

```bash
python3 hello.py
```

<b> Conda: </b>

```bash
conda activate py-practice
python hello.py
```

<b> uv: </b>

```bash
uv run python hello.py
```

⚠️ <strong>Note:</strong>
If you run your python file in a same folder, then above commands will work.  
However, if you are on a different folder in your terminal, then you need to pass correct path to execute a python file.

---

## Method 3. Run Python In VS Code

Prerequisite: VS Code is installed.

1. Install the Microsoft Python extension.
2. Open Command Palette with `Ctrl + Shift + P` or `Cmd + Shift + P` on macOS.
3. Search for `Python: Select Interpreter`.
4. Choose your Python environment.
5. Create `hello.py` and run it.

- Conda, choose your environment name.  

- uv, choose the interpreter inside the project `.venv`.  
macOS / Linux example: {project name}/.venv/bin/python  
Windows example: {project name}\.venv\Scripts\python.exe

---

## Method 4. Run Python In Jupyter Notebook

Jupyter Notebook is a browser-based environment where you write and run Python in small blocks called cells.

### With Anaconda

1. Open Anaconda Navigator.
2. Launch Jupyter Notebook or JupyterLab.
3. Create a new Python notebook.
4. Run a cell with `Shift + Enter`.

### With Conda

```bash
conda create -n py-practice python=3.12
conda activate py-practice
conda install jupyter
jupyter notebook
```

For JupyterLab:

```bash
jupyter lab
```

### With uv

```bash
uv init py-practice
cd py-practice
uv add --dev jupyter ipykernel
uv run jupyter notebook
```

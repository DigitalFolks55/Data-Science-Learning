# 1. Boolean

## 1) What is A "Boolean"

A <strong>Boolean</strong> represents one of two possible states, True or False.   
In python, boolean is recognized as binary number, 0 and 1:

| Value | Meaning |
|---|---|
| `True` | Yes / correct / condition is satisfied |
| `False` | No / incorrect / condition is not satisfied |

Boolean values are used heavily in:

| Situation | Example |
|---|---|
| Checking a condition | `age >= 20` |
| Controlling an `if` statement | `if is_adult:` |
| Filtering data | keep rows where a condition is `True` |
| Validating input | check whether a value is acceptable |

⚠️ <strong>Important:</strong> Python uses capitalized Boolean literals: `True` and `False`. `true` and `false` are not valid Boolean literals in Python.

````python
# Boolean values
is_student = True
is_teacher = False

# These are not boolean values
# is_student = true
# is_teacher = false

# Press shift+enter to run the code
````

````python
# Example: 1) What is A "Boolean"
print(is_student)
print(is_teacher)
print(type(is_student))
print(type(is_teacher))

# Press shift+enter to run the code
````

You can change data type to bool by using `bool`. In case, Zero would be `False` and others would be `True`

````python
# True
print(1, "is", bool(1))  # True
print(-1, "is", bool(-1)) # True
print("Hello", "is", bool("Hello"))  # True
print([1, 2, 3], "is", bool([1, 2, 3]))  # True
print({"key": "value"}, "is", bool({"key": "value"}), "\n")  # True

# False
print(0, "is", bool(0))  # False
print("", "is", bool(""))  # False
print([], "is", bool([]))  # False
print({}, "is", bool({}))  # False
print(None, "is", bool(None))  # False

# Press shift+enter to run the code
````

### 🗓️ Boolean is a built-in Python data type

A Boolean object has:

| Property | Example |
|---|---|
| Value | `True` |
| Type | `bool` |
| Identity | result of `id(True)` |

Boolean values are also related to numbers internally:

| Boolean | Numeric behavior |
|---|---:|
| `True` | behaves like `1` in arithmetic |
| `False` | behaves like `0` in arithmetic |

⚠️ <strong>Note:</strong> Even though this numeric behavior exists, use Boolean values mainly for conditions, not for normal numeric calculation.

````python
# Example: Boolean is a built-in Python data type
print("True as int:", int(True))
print("False as int:", int(False))

print("True + True:", True + True)
print("True + False:", True + False)
print("False + False:", False + False)

print("Is True an int?", isinstance(True, int))
print("Type of True:", type(True))

# Press shift+enter to run the code
````

````python
# Example: Boolean is a built-in Python data type
help(bool)

# Press shift+enter to run the code
````

## 2) Comparison Operators

A comparison operator compares two values and returns a Boolean result.

| Operator | Meaning | Example |
|---|---|---|
| `==` | equal to | `x == y` |
| `!=` | not equal to | `x != y` |
| `>` | greater than | `x > y` |
| `<` | less than | `x < y` |
| `>=` | greater than or equal to | `x >= y` |
| `<=` | less than or equal to | `x <= y` |

⚠️ <strong>Watch out:</strong> `=` means assignment. `==` means equality comparison.

````python
# Example: 2) Comparison Operators
x = 10
y = 3

print("x == y:", x == y)
print("x != y:", x != y)
print("x > y:", x > y)
print("x < y:", x < y)
print("x >= y:", x >= y)
print("x <= y:", x <= y)

# Press shift+enter to run the code
````

### 🗓️ Assignment Versus Comparison

"=": Assignment  
"==": Comparison (Equal or not?)

| Code | Meaning |
|---|---|
| `score = 80` | Assign `80` to variable `score` |
| `score == 80` | Check whether `score` is equal to `80` |

In Python, `if score = 80:` causes an error because assignment is not a condition.

````python
# Assignment of a literal value to a variable
score = 80

# Comparison of the variable with a literal value
print("score:", score)
print("score == 80:", score == 80)
print("score == 100:", score == 100)

# Press shift+enter to run the code
````

### 🗓️ Chained Comparisons

Python can chain comparisons in a natural mathematical style.

```python
0 <= score <= 100
```

This means:

```python
0 <= score and score <= 100
```

````python
# Example: Chained Comparisons
score = 85

print("0 <= score <= 100:", 0 <= score <= 100)
print("0 <= score and score <= 100:", 0 <= score and score <= 100)

score = 120
print("0 <= score <= 100:", 0 <= score <= 100)

# Press shift+enter to run the code
````

## 3) Logical Operators

Logical operators combine Boolean expressions.

| Operator | Meaning | Example |
|---|---|---|
| `and` | `True` only if both sides are `True` | `age >= 20 and has_ticket` |
| `or` | `True` if at least one side is `True` | `is_student or is_teacher` |
| `not` | reverses Boolean value | `not is_admin` |

Use logical operators when one condition is not enough.

````python
# Example: 3) Logical Operators
age = 22
has_ticket = True

print("Can enter?", age >= 20 and has_ticket)

is_student = False
is_teacher = True
print("Can access classroom?", is_student or is_teacher)

is_admin = False
print("Not admin?", not is_admin)

# Press shift+enter to run the code
````

### 🗓️ Truth Table

| A | B | A and B | A or B |
|---|---|---|---|
| `True` | `True` | `True` | `True` |
| `True` | `False` | `False` | `True` |
| `False` | `True` | `False` | `True` |
| `False` | `False` | `False` | `False` |

❓ <strong>Discussion:</strong> For a login system, which one is safer: `username_is_valid and password_is_valid` or `username_is_valid or password_is_valid`?

````python
# Example: Truth Table
cases = [
    (True, True),
    (True, False),
    (False, True),
    (False, False),
]

for a, b in cases:
    print(f"A={a:5} B={b:5} | A and B = {a and b:5} | A or B = {a or b:5}")

# Press shift+enter to run the code
````

### 🗓️ Operator Precedence

`not`, `and`, and `or` have different priority.

| Priority | Operator |
|---:|---|
| 1 | `not` |
| 2 | `and` |
| 3 | `or` |

When conditions become complex, use parentheses `()` to make your intention clear.

````python
# Example: Operator Precedence
a = True
b = False
c = False

print("Without parentheses:", a or b and c)
print("With parentheses 1:", (a or b) and c)
print("With parentheses 2:", a or (b and c))

# Press shift+enter to run the code
````

## 4) Truthy And Falsy Values

Python can convert many objects to Boolean values.

Some values are treated as `False`. These are called <strong>falsy</strong> values.

| Falsy value | Example |
|---|---|
| zero integer | `0` |
| zero float | `0.0` |
| empty string | `""` |
| empty list | `[]` |
| empty tuple | `()` |
| empty dictionary | `{}` |
| empty set | `set()` |
| `None` | `None` |

Most other values are treated as `True`. These are called <strong>truthy</strong> values.

````python
# Example: 4) Truthy And Falsy Values
values = [0, 1, -1, 0.0, 0.1, "", "Python", "False", [], [1, 2], {}, {"a": 1}, None]

for value in values:
    print(f"value={repr(value):>12} | bool(value)={bool(value)}")

# Press shift+enter to run the code
````

# 2. Numbers

Python has several numeric types.

| Type | Name | Example | Use case |
|---|---|---|---|
| `int` | integer | `100` | Whole numbers |
| `float` | floating-point number | `3.14` | Decimal numbers / approximation |
| `complex` | complex number | `2 + 3j` | Advanced math, signal processing, scientific computing |
| `radix` | base number | `0b10` (10 in decimal ) | base 2 for binary valuesor base 16 for hexadecimal values used by computers |

## 1) Integer: `int`

An integer is a whole number without decimal points.

Examples: 0, 1, -10, 1000000

- Invalid literal integer: start with 0 for 1-9.
- We can put plus or minus at the begining of integer; +123, -123
- Separator is not "," but "_"; 1,000 (X) 1_000 (O)

````python
# Example: 1) Integer: `int`
small_number = 10
large_number = 10 ** 50
negative_number = -7
amount_with_underscores = 1_000
amount_with_comma = 1,000  # This will be a tuple, not an integer

print(small_number)
print(large_number)
print(negative_number)
print(amount_with_underscores)
print(amount_with_comma)

# Press shift+enter to run the code
````

Python integers can be very large. They are not limited to 32-bit or 64-bit size in normal Python code.


## 2) Float: `float`

A float is a number with a decimal point.

Examples: 3.14, 0.5, -2.0, 1e6

`1e6` means `1 × 10^6` (scientific number), which is `1000000.0`.

````python
# Example: 2) Float: `float`
pi = 3.14
half = 0.5
scientific = 1e6

print(pi, type(pi))
print(half, type(half))
print(scientific, type(scientific))

# Press shift+enter to run the code
````

## 3) Complex: `complex`

A complex number has a real part and an imaginary part.

Python uses `j` for the imaginary unit.

```python
z = 2 + 3j
```

| Attribute | Meaning |
|---|---|
| `z.real` | real part |
| `z.imag` | imaginary part |

````python
# Example: 3) Complex: `complex`
z = 2 + 3j

print("z:", z)
print("type:", type(z))
print("real part:", z.real)
print("imaginary part:", z.imag)

# Press shift+enter to run the code
````

## 4) Radix

A **radix** is the **base** of a number system.

In Python, radix indicates how a number should be interpreted.

For example:

- Binary uses radix `2`
- Octal uses radix `8`
- Decimal uses radix `10`
- Hexadecimal uses radix `16`

Python uses prefixes to represent non-decimal numbers:

- `0b` for binary
- `0o` for octal
- `0x` for hexadecimal

````python
# Example: 4) Radix
binary_number = 0b1010  # 10 in decimal
octal_number = 0o12     # 10 in decimal
hexadecimal_number = 0xA  # 10 in decimal

print("Binary:", binary_number)
print("Octal:", octal_number)
print("Hexadecimal:", hexadecimal_number)

# Press shift+enter to run the code
````

## 5) Arithmetic Operators

Python supports common arithmetic operators.

| Operator | Meaning | Example | Result | Note |
|---|---|---|---|---|
| `+` | addition | `10 + 3` | `13` | |
| `-` | subtraction | `10 - 3` | `7` | |
| `*` | multiplication | `10 * 3` | `30` | |
| `/` | true division | `10 / 3` | `3.333...` | Always returns a `float` | 
| `//` | floor division | `10 // 3` | `3` | Returns the floor value |
| `%` | modulo / remainder | `10 % 3` | `1` | |
| `**` | exponentiation | `10 ** 3` | `1000` | |

⚠️ <strong>Important:</strong> Floor division means round down toward negative infinity, not simply remove the decimal part.

````python
# Assignment of a literal value to a variable
x = 10
y = 3

print("Arithmetic operations")
print("x + y  =", x + y)
print("x - y  =", x - y)
print("x * y  =", x * y)
print("x / y  =", x / y)
print("x // y =", x // y)
print("x % y  =", x % y)
print("x ** y =", x ** y)
print()

# Data type of the result of division
print("Data type of the result of division")
print("10 / 3  =", 10 / 3, type(10 / 3))
print("10 // 3 =", 10 // 3, type(10 // 3))

print("-10 / 3  =", -10 / 3, type(-10 / 3))
print("-10 // 3 =", -10 // 3, type(-10 // 3))

# Press shift+enter to run the code
````

### 🗓️ Assignment of result of arithmetic operation
If you don't assign calculated values to original or new variable, the value will not be updated.  
For example, you assign a = 2 and add 1 to a.  

```python
a = 2
a + 1
```

Do you think a will be 3 or 2?  
In this case, the value of a will not be updated.  
There are two approaches to update values.
```python
a = 2
a = a + 1  # Assign object again
a += 1  # Use arithmetic operator with assignment
```


````python
# Wrong case
a = 3  # assign 3 to a
print("Initial value of a:", a)
a + 2  # add 2 to a
print("Try a + 2")
print("New value of a:", a)
print("Value has been updated? \n")

# Assign object again
a = 3
print("Initial value of a:", a)
a = a + 2  # add 2 to a and assign back to a
print("Try a = a + 2")
print("New value of a:", a)
print("Value has been updated? \n")

# Arithmetic operaotors with assignment
a = 3
print("Initial value of a:", a)
a += 2  # equivalent to a = a + 2
print("Try a += 2")
print("New value of a:", a) 
print("Value has been updated? \n")

# Press shift+enter to run the code
````

## 6) Numeric Type Conversion

You can convert values between numeric types.

| Function | Meaning | Example |
|---|---|---|
| `int(x)` | convert to integer | `int("10")` |
| `float(x)` | convert to float | `float("3.14")` |
| `complex(x)` | convert to complex | `complex(10)` |
| `bin(x)` | convert to binary | `bin(1)` |
| `oct(x)` | convert to octal | `oct(1)` |
| `hex(x)` | convert to hexadecimal | `hex(1)` |

⚠️ <strong>Watch out:</strong> Conversion can fail if the text is not a valid number.

````python
# Example: 6) Numeric Type Conversion
print(int("10"))
print(float("3.14"))
print(complex(10.2))
print(bin(10))
print(oct(10))
print(hex(10))

# Press shift+enter to run the code
````

### 🗓️ `int()` Does Not Round

`int()` removes the decimal part toward zero.

For rounding, use `round()`.

````python
# Example: `int()` Does Not Round
print("int(2.9):", int(2.9))
print("int(-2.9):", int(-2.9))

print("round(2.9):", round(2.9))
print("round(-2.9):", round(-2.9))

# Press shift+enter to run the code
````

### 🗓️ Safe Conversion Pattern

When user input or external data may contain invalid values, use `try` and `except`.  
If not, you may get errors.

````python
# Example: Safe Conversion Pattern
texts = ["10", "3.14", "hello", ""]

for text in texts:
    try:
        number = float(text)
        print(f"{repr(text):>8} -> {number}")
    except ValueError:
        print(f"{repr(text):>8} -> cannot convert to float")

# Press shift+enter to run the code
````

### ⚠️ <strong> Floating-Point Caution </strong>

Floating-point numbers are approximations. Some decimal numbers cannot be represented exactly in binary.

This can produce surprising results.

````python
# Example: Floating-Point Caution
print("0.1 + 0.2 =", 0.1 + 0.2)
print("0.1 + 0.2 == 0.3:", 0.1 + 0.2 == 0.3)

# Press shift+enter to run the code
````

You can use a tolerance instead of exact equality to compare floats.  
Python has `math.isclose()` for this purpose.

````python
# Example: Floating-Point Caution
import math

x = 0.1 + 0.2
y = 0.3

print("x:", x)
print("y:", y)
print("x == y:", x == y)
print("math.isclose(x, y):", math.isclose(x, y))

# Press shift+enter to run the code
````

### 🗓️ Exact Decimal Calculation

For money-like decimal calculation, use `Decimal` from the `decimal` module.

⚠️ <strong>Important:</strong> Pass strings such as `"0.1"` into `Decimal`, not floats such as `0.1`.

````python
# Example: Exact Decimal Calculation
from decimal import Decimal

x = Decimal("0.1") + Decimal("0.2")
y = Decimal("0.3")

print(x)
print(y)
print(x == y)

# Press shift+enter to run the code
````

# 4. Key Terms

| Term | Meaning |

| Boolean | A data type with only two values: `True` and `False` |

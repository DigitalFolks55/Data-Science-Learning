# 1. Conditions

## 1) What Is A Condition?

A <strong>condition</strong> is an expression that Python evaluates as `True` or `False`.

Conditions are used when your program needs to make a decision.

| Example | Result type | Meaning |
|---|---|---|
| `score >= 60` | `bool` | Is the score passing? |
| `name == "Alice"` | `bool` | Is the name Alice? |
| `"py" in "python"` | `bool` | Does the text contain `"py"`? |
| `bool(items)` | `bool` | Does the collection have any items? |

````python
# Example: check simple conditions
score = 75
name = "Alice"
items = ["notebook", "pen"]

print(score >= 60)
print(name == "Alice")
print("py" in "python")
print(bool(items))

# Press shift+enter to run the code
````

## 2) Comparison Operators

Comparison operators compare two values and return `True` or `False`.

| Operator | Meaning | Example |
|---|---|---|
| `==` | Equal to | `score == 100` |
| `!=` | Not equal to | `status != "done"` |
| `<` | Less than | `age < 18` |
| `<=` | Less than or equal to | `score <= 60` |
| `>` | Greater than | `price > 1000` |
| `>=` | Greater than or equal to | `score >= 60` |

⚠️ <strong>Important:</strong> Use `==` to compare values. Use `=` only when assigning a value to a variable.

````python
# Example: comparison operators
score = 75
target_score = 80

print(score == target_score)
print(score != target_score)
print(score < target_score)
print(score <= target_score)
print(score > target_score)
print(score >= 60)

# Press shift+enter to run the code
````

## 3) `if` Statement

An `if` statement runs a block of code only when the condition is `True`.

Basic syntax:

```python
if condition:
    code_to_run
```

Python uses indentation to know which code belongs inside the `if` block.

````python
# Example: run code only when the condition is True
score = 82

if score >= 60:
    print("Passed")

print("This line runs regardless of the if condition.")

# Press shift+enter to run the code
````

## 4) `if-else`

Use `else` when you want one block to run when the condition is `True`, and another block to run when the condition is `False`.

````python
# Example: choose between two outcomes
score = 45

if score >= 60:
    print("Passed")
else:
    print("Try again")

# Press shift+enter to run the code
````

## 5) `if-elif-else`

Use `elif` when you need multiple conditions.

Python checks conditions from top to bottom and runs the first block whose condition is `True`.

````python
# Example: assign a grade using if-elif-else
score = 86

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "Needs improvement"

print("Grade:", grade)

# Press shift+enter to run the code
````

## 6) What Is True?

Python can treat many values as `True` or `False` in conditions.

| Falsy values | Examples |
|---|---|
| Zero numbers | `0`, `0.0` |
| Empty collections | `""`, `[]`, `()`, `{}`, `set()` |
| Missing value | `None` |
| Boolean false | `False` |

Most other values are <strong>truthy</strong>.

````python
# Example: truthy and falsy values in if statements
values = [0, 1, "", "Python", [], [1, 2], None]

for value in values:
    if value:
        print(repr(value), "is truthy")
    else:
        print(repr(value), "is falsy")

# Press shift+enter to run the code
````

## 7) Multiple Conditions With `and`, `or`, And `not`

Logical operators combine conditions.

| Operator | Meaning |
|---|---|
| `and` | `True` only when both conditions are true |
| `or` | `True` when at least one condition is true |
| `not` | Reverses a Boolean value |

````python
# Example: combine multiple conditions
age = 22
has_ticket = True
is_blocked = False

if age >= 18 and has_ticket and not is_blocked:
    print("Can enter")
else:
    print("Cannot enter")

# Press shift+enter to run the code
````

## 8) Using `in` For Multiple Conditions

When one value should match one of several options, `in` is often cleaner than many `or` conditions.

Instead of:

```python
role == "admin" or role == "staff" or role == "owner"
```

Use:

```python
role in ["admin", "staff", "owner"]
```

````python
# Example: use in for multiple accepted values
role = "staff"
allowed_roles = ["admin", "staff", "owner"]

if role in allowed_roles:
    print("Access granted")
else:
    print("Access denied")

status = "pending"

if status not in ["cancelled", "blocked"]:
    print("Process can continue")

# Press shift+enter to run the code
````

## 9) Assignment Expression `:=`

The assignment expression operator `:=` is also called the <strong>walrus operator</strong>.

It assigns a value and checks it in the same expression.

Basic idea:

```python
if variable := expression:
    use variable
```

Use it carefully. It can make short code clearer, but overusing it can make code harder to read.

````python
# Example: assign and check a value in one condition
message = "Python"

if length := len(message):
    print("Message length:", length)
else:
    print("Message is empty")

names = ["Alice", "Ben", "Chloe"]

if first_name := names[0]:
    print("First name:", first_name)

# Press shift+enter to run the code
````

## 10) Conditional Expression

A conditional expression is a compact `if-else` expression.

Syntax:

```python
value_if_true if condition else value_if_false
```

Use it for simple choices only.

````python
# Example: conditional expression
score = 72
result = "Passed" if score >= 60 else "Failed"

print(result)

# Press shift+enter to run the code
````

# 2. While Loops

## 1) What Is A `while` Loop?

A `while` loop repeats while its condition is `True`.

Basic syntax:

```python
while condition:
    code_to_repeat
```

⚠️ <strong>Important:</strong> Make sure the condition eventually becomes `False`, or the loop can run forever.

````python
# Example: repeat while a condition is True
count = 1

while count <= 3:
    print("Count:", count)
    count += 1

# Press shift+enter to run the code
````

## 2) `break` In A `while` Loop

`break` stops the loop immediately.

````python
# Example: stop a while loop with break
count = 1

while count <= 10:
    if count == 4:
        print("Stop at", count)
        break
    print(count)
    count += 1

# Press shift+enter to run the code
````

## 3) `continue` In A `while` Loop

`continue` skips the rest of the current loop cycle and moves to the next cycle.

````python
# Example: skip one value with continue
count = 0

while count < 5:
    count += 1
    if count == 3:
        continue
    print(count)

# Press shift+enter to run the code
````

## 4) `while ... else`

A `while` loop can have an `else` block.

The `else` block runs only if the loop finishes naturally without `break`.

````python
# Example: while-else when no break happens
target = 7
number = 1

while number <= 5:
    if number == target:
        print("Found target")
        break
    number += 1
else:
    print("Target was not found")

# Press shift+enter to run the code
````

````python
# Example: while-else skipped because break happens
target = 3
number = 1

while number <= 5:
    if number == target:
        print("Found target")
        break
    number += 1
else:
    print("Target was not found")

# Press shift+enter to run the code
````

## 5) `pass` In A `while` Loop

`pass` means "do nothing for now." It is useful as a placeholder when Python requires a block of code.

````python
# Example: pass as a placeholder
count = 0

while count < 3:
    count += 1
    if count == 2:
        pass
    print("Count:", count)

# Press shift+enter to run the code
````

# 3. For Loops

## 1) What Is A `for` Loop?

A `for` loop repeats once for each item in an iterable.

Common iterables include strings, lists, tuples, dictionaries, sets, and ranges.

````python
# Example: loop through a list
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print(fruit)

# Press shift+enter to run the code
````

## 2) Looping Through Different Data Types

The same `for ... in` pattern works with many Python objects.

````python
# Example: iterate through different iterable objects
for character in "Python":
    print("Character:", character)

scores = {"Alice": 90, "Ben": 82}

for name, score in scores.items():
    print(name, score)

# Press shift+enter to run the code
````

## 3) `break` In A `for` Loop

`break` stops the loop immediately.

````python
# Example: stop a for loop with break
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    if number == 4:
        print("Stop at", number)
        break
    print(number)

# Press shift+enter to run the code
````

## 4) `continue` In A `for` Loop

`continue` skips the rest of the current loop cycle and moves to the next item.

````python
# Example: skip even numbers with continue
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    if number % 2 == 0:
        continue
    print(number)

# Press shift+enter to run the code
````

## 5) `for ... else`

A `for` loop can have an `else` block.

The `else` block runs only if the loop finishes naturally without `break`.

````python
# Example: for-else when no break happens
names = ["Alice", "Ben", "Chloe"]
target = "David"

for name in names:
    if name == target:
        print("Found", target)
        break
else:
    print(target, "was not found")

# Press shift+enter to run the code
````

````python
# Example: for-else skipped because break happens
names = ["Alice", "Ben", "Chloe"]
target = "Ben"

for name in names:
    if name == target:
        print("Found", target)
        break
else:
    print(target, "was not found")

# Press shift+enter to run the code
````

## 6) `pass` In A `for` Loop

Use `pass` as a placeholder when a loop body is not ready yet.

````python
# Example: pass as a placeholder in a for loop
tasks = ["clean", "transform", "load"]

for task in tasks:
    if task == "transform":
        pass
    print("Task:", task)

# Press shift+enter to run the code
````

# 4. Sequence Tools

## 1) `range()`

`range()` creates a sequence of numbers, often used with `for` loops.

Common patterns:

| Code | Meaning |
|---|---|
| `range(5)` | `0, 1, 2, 3, 4` |
| `range(2, 6)` | `2, 3, 4, 5` |
| `range(2, 10, 2)` | `2, 4, 6, 8` |

````python
# Example: create number sequences with range
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(2, 10, 2)))

for number in range(3):
    print("Loop number:", number)

# Press shift+enter to run the code
````

## 2) `enumerate()`

`enumerate()` gives both index and value while looping.

````python
# Example: get index and value with enumerate
fruits = ["apple", "banana", "orange"]

for index, fruit in enumerate(fruits):
    print(index, fruit)

# Press shift+enter to run the code
````

## 3) `zip()`

`zip()` loops through multiple iterables together.

⚠️ <strong>Note:</strong> `zip()` stops when the shortest iterable ends.

````python
# Example: combine related lists with zip
names = ["Alice", "Ben", "Chloe"]
scores = [90, 82, 95]

for name, score in zip(names, scores):
    print(name, score)

# Press shift+enter to run the code
````

## 4) `reversed()` And `sorted()`

`reversed()` iterates in reverse order.  
`sorted()` returns a sorted list from an iterable.

````python
# Example: reversed and sorted
numbers = [3, 1, 4, 2]

print("Original:", numbers)
print("Reversed:", list(reversed(numbers)))
print("Sorted:", sorted(numbers))
print("Original after sorted():", numbers)

# Press shift+enter to run the code
````

## 5) Useful `itertools` Helpers

`itertools` is a standard Python library for efficient iteration patterns.

Useful helpers include:

| Helper | Meaning |
|---|---|
| `count()` | Infinite counting sequence |
| `cycle()` | Repeat values forever |
| `repeat()` | Repeat one value |
| `islice()` | Take part of an iterator |
| `chain()` | Combine iterables into one stream |
| `product()` | Cartesian product |
| `combinations()` | Choose items without order |
| `permutations()` | Arrange items with order |

````python
# Example: useful itertools helpers
from itertools import count, cycle, repeat, islice, chain, product, combinations, permutations

print("count:", list(islice(count(10), 5)))
print("cycle:", list(islice(cycle(["A", "B"]), 6)))
print("repeat:", list(repeat("Python", 3)))
print("chain:", list(chain([1, 2], [3, 4])))
print("product:", list(product(["A", "B"], [1, 2])))
print("combinations:", list(combinations(["A", "B", "C"], 2)))
print("permutations:", list(permutations(["A", "B", "C"], 2)))

# Press shift+enter to run the code
````

## 6) Other Helpful Standard Library Tools

Python also has standard library tools that pair well with loops.

| Tool | Use case |
|---|---|
| `collections.Counter` | Count how often values appear |
| `collections.defaultdict` | Group values without manual key setup |
| `operator.itemgetter` | Sort or select structured data by position/key |

````python
# Example: Counter and defaultdict with loops
from collections import Counter, defaultdict
from operator import itemgetter

words = ["python", "sql", "python", "excel", "sql", "python"]
print("Counter:", Counter(words))

records = [
    {"department": "Sales", "name": "Alice"},
    {"department": "Data", "name": "Ben"},
    {"department": "Sales", "name": "Chloe"},
]

groups = defaultdict(list)
for record in records:
    groups[record["department"]].append(record["name"])

print("Groups:", dict(groups))
print("Sorted by department:", sorted(records, key=itemgetter("department")))

# Press shift+enter to run the code
````

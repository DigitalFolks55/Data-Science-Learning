# 1. List

## 1) What Is A List?

A <strong>list</strong> stores multiple values in one variable.

Lists are:

| Feature | Meaning |
|---|---|
| Ordered | Items keep their position |
| Mutable | You can change the list after creation |
| Allow duplicates | The same value can appear more than once |
| Mixed type capable | A list can contain different data types |

Lists are useful when you need a collection that can grow, shrink, or change.

````python
fruits = ["apple", "banana", "orange"]
numbers = [10, 20, 30]
mixed_values = ["Alice", 25, True, 82.5]

print(fruits)
print(numbers)
print(mixed_values)
print(type(fruits))
````

### 🗓️ List Is An Object

A list object has:

| Property | Example |
|---|---|
| Value | `["apple", "banana"]` |
| Type | `list` |
| Identity | result of `id(fruits)` |
| Methods | `.append()`, `.remove()`, `.sort()`, and many more |

Because lists are objects, you can inspect them with built-in functions.

````python
fruits = ["apple", "banana", "orange"]

print("Value:", fruits)
print("Type:", type(fruits))
print("Identity:", id(fruits))
print("Length:", len(fruits))
print("Available methods include:")
print([name for name in dir(fruits) if not name.startswith("_")])
````

---

## 2) Indexing And Slicing Lists

Lists are ordered, so each item has an index.

For `["apple", "banana", "orange"]`:

| Item | apple | banana | orange |
|---|---:|---:|---:|
| Positive index | 0 | 1 | 2 |
| Negative index | -3 | -2 | -1 |

Python indexing starts from `0`.

````python
fruits = ["apple", "banana", "orange"]

print(fruits[0])
print(fruits[1])
print(fruits[-1])
print(fruits[-2])
````

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[0:3])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])
````

### 🗓️ IndexError

If the index does not exist, Python raises `IndexError`.

```python
fruits = ["apple", "banana", "orange"]
fruits[100]
```

Keep this in mind when using indexes from user input or external data.

---

## 3) Updating List Items

Lists are mutable, so you can change an item by assigning a new value to an index.

````python
fruits = ["apple", "banana", "orange"]

fruits[1] = "grape"

print(fruits)
````

### 🗓️ Adding Items

Common ways to add items:

| Method | Meaning |
|---|---|
| `.append(x)` | Add one item to the end |
| `.insert(index, x)` | Add one item at a specific position |
| `.extend(items)` | Add multiple items from another iterable |

````python
fruits = ["apple", "banana"]

fruits.append("orange")
fruits.insert(1, "grape")
fruits.extend(["melon", "kiwi"])

print(fruits)
````

### 🗓️ Removing Items

Common ways to remove items:

| Method | Meaning |
|---|---|
| `.remove(x)` | Remove the first matching value |
| `.pop()` | Remove and return the last item |
| `.pop(index)` | Remove and return the item at an index |
| `.clear()` | Remove all items |

````python
fruits = ["apple", "banana", "orange", "banana"]

fruits.remove("banana")
last_fruit = fruits.pop()

print("List:", fruits)
print("Popped item:", last_fruit)
````

---

## 4) Sorting And Reversing Lists

Lists can be sorted or reversed.

| Tool | Meaning |
|---|---|
| `.sort()` | Sort the original list in place |
| `sorted(list)` | Return a new sorted list |
| `.reverse()` | Reverse the original list in place |
| `reversed(list)` | Return an iterator in reverse order |

⚠️ <strong>Important:</strong> `.sort()` changes the original list. `sorted()` creates a new sorted result.

````python
scores = [82, 95, 71, 88]

sorted_scores = sorted(scores)

print("Original:", scores)
print("Sorted result:", sorted_scores)

scores.sort(reverse=True)
print("Original after .sort(reverse=True):", scores)
````

### 🗓️ List Membership And Counting

Use `in` to check membership. Use `.count()` to count occurrences.

````python
fruits = ["apple", "banana", "orange", "banana"]

print("apple" in fruits)
print("grape" in fruits)
print("banana count:", fruits.count("banana"))
print("orange index:", fruits.index("orange"))
````

---

## 5) List Copying

Assignment does not copy a list. It gives another name to the same list object.

```python
a = [1, 2, 3]
b = a
```

Now `a` and `b` refer to the same list.

````python
a = [1, 2, 3]
b = a

b.append(4)

print("a:", a)
print("b:", b)
print("Same object?", a is b)
````

````python
a = [1, 2, 3]
b = a.copy()

b.append(4)

print("a:", a)
print("b:", b)
print("Same object?", a is b)
````

### 🗓️ Nested Lists

A nested list is a list inside another list. Nested lists are common in table-like data.

````python
grades = [
    ["Alice", 90],
    ["Ben", 82],
    ["Chloe", 95],
]

print(grades[0])
print(grades[0][0])
print(grades[0][1])
````

# 2. Tuple

## 1) What Is A Tuple?

A <strong>tuple</strong> stores multiple values in one variable, like a list, but it is immutable.

Tuples are:

| Feature | Meaning |
|---|---|
| Ordered | Items keep their position |
| Immutable | You cannot change items after creation |
| Allow duplicates | The same value can appear more than once |
| Good for fixed records | Useful for values that should stay together |

````python
point = (10, 20)
person = ("Alice", 25, "Data Analyst")
single_item_tuple = ("Python",)

print(point)
print(person)
print(single_item_tuple)
print(type(point))
print(type(single_item_tuple))
````

### 🗓️ Single-Item Tuple

A single-item tuple needs a comma.

| Code | Type |
|---|---|
| `("Python")` | `str` |
| `("Python",)` | `tuple` |

````python
not_tuple = ("Python")
real_tuple = ("Python",)

print(type(not_tuple))
print(type(real_tuple))
````

---

## 2) Tuple Indexing And Unpacking

Tuples support indexing and slicing like lists.

Tuple unpacking assigns tuple items to multiple variables.

````python
point = (10, 20, 30)

print(point[0])
print(point[-1])
print(point[0:2])

x, y, z = point
print(x, y, z)
````

### 🗓️ Tuple Immutability

Tuples cannot be changed in place.

```python
point = (10, 20)
point[0] = 99  # TypeError
```

If you need to change items often, use a list instead.

````python
point = (10, 20)

# This would cause TypeError if uncommented.
# point[0] = 99

new_point = (99, point[1])
print("Original:", point)
print("New:", new_point)
````

## 3) Tuple Methods

Tuples have fewer methods than lists because they cannot be changed.

| Method | Meaning |
|---|---|
| `.count(x)` | Count how many times a value appears |
| `.index(x)` | Return the index of the first matching value |

````python
colors = ("red", "blue", "red", "green")

print(colors.count("red"))
print(colors.index("green"))
````

# 3. Set

## 1) What Is A Set?

A <strong>set</strong> stores unique values.

Sets are:

| Feature | Meaning |
|---|---|
| Unordered | Items do not keep a fixed position |
| Mutable | You can add and remove items |
| Unique | Duplicate values are removed |
| Fast for membership checks | Good for checking whether something exists |

Sets are useful when you care about uniqueness, not order.

````python
numbers = {1, 2, 2, 3, 3, 3}
skills = {"Python", "SQL", "Python"}

print(numbers)
print(skills)
print(type(skills))
````

### 🗓️ Empty Set

Be careful when creating an empty set.

| Code | Type |
|---|---|
| `{}` | `dict` |
| `set()` | `set` |

````python
empty_dict = {}
empty_set = set()

print(type(empty_dict))
print(type(empty_set))
````

---

## 2) Adding, Removing, And Checking Set Items

Common set methods:

| Method | Meaning |
|---|---|
| `.add(x)` | Add one item |
| `.update(items)` | Add multiple items |
| `.remove(x)` | Remove item; error if missing |
| `.discard(x)` | Remove item; no error if missing |
| `.clear()` | Remove all items |

````python
skills = {"Python", "SQL"}

skills.add("Excel")
skills.update(["Pandas", "Python"])
skills.discard("Tableau")

print(skills)
print("Python" in skills)
print("Java" in skills)
````

### 🗓️ `remove()` Versus `discard()`

`remove()` raises an error when the item does not exist.  
`discard()` does nothing when the item does not exist.

Use `discard()` when missing values are acceptable.

````python
skills = {"Python", "SQL"}

skills.discard("Java")
print(skills)

# This would cause KeyError if uncommented.
# skills.remove("Java")
````

---

## 3) Set Operations

Sets are useful for comparing groups.

| Operation | Python syntax | Meaning |
|---|---|---|
| Union | `a | b` or `a.union(b)` | Items in either set |
| Intersection | `a & b` or `a.intersection(b)` | Items in both sets |
| Difference | `a - b` or `a.difference(b)` | Items in `a` but not `b` |
| Symmetric difference | `a ^ b` | Items in either set, but not both |

````python
python_students = {"Alice", "Ben", "Chloe"}
sql_students = {"Ben", "David", "Chloe"}

print("Union:", python_students | sql_students)
print("Intersection:", python_students & sql_students)
print("Python only:", python_students - sql_students)
print("One course only:", python_students ^ sql_students)
````

### 🗓️ Removing Duplicates With Set

You can convert a list to a set to remove duplicates.

⚠️ <strong>Important:</strong> A set does not preserve the original order as a learning rule. If order matters, use another approach later.

````python
raw_tags = ["python", "sql", "python", "excel", "sql"]
unique_tags = set(raw_tags)

print(raw_tags)
print(unique_tags)
````

# 5. Key Terms

| Term | Meaning |
|---|---|
| Collection | A data type that can hold multiple values |
| List | An ordered, mutable collection written with square brackets `[]` |
| Tuple | An ordered, immutable collection written with parentheses `()` |
| Set | An unordered collection of unique values written with `{}` or created with `set()` |
| Ordered | Items keep a position that can be accessed by index |
| Unordered | Items do not have a reliable position for indexing |
| Mutable | Can be changed after creation |
| Immutable | Cannot be changed after creation |
| Index | A position inside an ordered collection |
| Slicing | Extracting part of an ordered collection with `start:stop:step` |
| Nested list | A list inside another list |
| Append | Add one item to the end of a list |
| Extend | Add multiple items to a list |
| Pop | Remove and return an item from a list |
| Tuple unpacking | Assigning tuple items to multiple variables at once |
| Unique values | Values that appear only once in a set |
| Membership check | Checking whether a value exists with `in` |
| Union | Set operation that combines items from both sets |
| Intersection | Set operation that keeps items found in both sets |
| Difference | Set operation that keeps items in one set but not another |

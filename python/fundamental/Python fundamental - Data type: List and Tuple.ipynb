# 1. List

## 1) What Is A List?

A <strong>list</strong> stores multiple values in one variable.

Lists are:

| Feature | Meaning |
|---|---|
| Ordered | Items keep their position |
| Mutable | You can change the list after creation |
| Allow duplicates | The same value can appear more than once |
| Mixed type capable | A list can contain different data types |

Lists are useful when you need a collection that can grow, shrink, or change.

````python
# Example: 1) What Is A List?
fruits = ["apple", "banana", "orange"]
numbers = [10, 20, 30]
mixed_values = ["Alice", 25, True, 82.5]

print(fruits)
print(numbers)
print(mixed_values)
print(type(fruits))

# Press shift+enter to run the code
````

### 🗓️ List Is An Object

A list object has:

| Property | Example |
|---|---|
| Value | `["apple", "banana"]` |
| Type | `list` |
| Identity | result of `id(fruits)` |
| Methods | `.append()`, `.remove()`, `.sort()`, and many more |

Because lists are objects, you can inspect them with built-in functions.

````python
# Example: List Is An Object
fruits = ["apple", "banana", "orange"]

print("Value:", fruits)
print("Type:", type(fruits))
print("Identity:", id(fruits))
print("Length:", len(fruits))  # len() returns the number of items in the list
print("Available methods include:")
print([name for name in dir(fruits) if not name.startswith("_")])

# Press shift+enter to run the code
````

## 2) How to create a list

Lists can be generated by

| Method | Meaning |
|---|---|
| `[]` | Generate an empty list or a list with elements |
| `list()` | Convert string, tuple, dictionary & set to list|
| `split()` | Convert string to list using separator |

````python
# Use []
empty_list = []
print("Empty list:", empty_list)
not_empty_list = [1, 2, 3]
print("Not empty list:", not_empty_list)

# Use list()
another_empty_list = list()
print("Another empty list:", another_empty_list)
another_not_empty_list = list([4, 5, 6])
print("Another not empty list:", another_not_empty_list)
string_list = list("Hello")
print("String to list:", string_list)
tuple_list = list((7, 8, 9))
print("Tuple to list:", tuple_list)
dict_list = list({"a": 1, "b": 2})
print("Dictionary to list (keys):", dict_list)
set_list = list({10, 20, 30})
print("Set to list:", set_list)

# Use split()
sentence = "Hello world, welcome to Python programming!"
words = sentence.split()
print("Words:", words)
data = "name,age,city\nAlice,30,New York\nBob,25,Los Angeles"
lines = data.split("\n")
print("Lines:", lines)

# Press shift+enter to run the code
````

## 3) Indexing And Slicing Lists

Lists are ordered, so each item has an index.

For `["apple", "banana", "orange"]`:

| Item | apple | banana | orange |
|---|---:|---:|---:|
| Positive index | 0 | 1 | 2 |
| Negative index | -3 | -2 | -1 |

Python indexing starts from `0`.

````python
# Example: 3) Indexing And Slicing Lists
fruits = ["apple", "banana", "orange"]

print(fruits[0])
print(fruits[1])
print(fruits[-1])
print(fruits[-2])

# Press shift+enter to run the code
````

````python
# Example: 3) Indexing And Slicing Lists
numbers = [10, 20, 30, 40, 50]

print(numbers[0:3])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])

# Press shift+enter to run the code
````

### 🗓️ Find index

`.index()` allows to find an index of element.

````python
# Example: Find index
students = ["Alice", "Bob", "Charlie", "David", "Eve"]
print(students.index("Charlie"))
print(students.count("Lee"))  # if the item is not found, returns 0

# Press shift+enter to run the code
````

## 4) Updating List Items

Lists are mutable, so you can change an item by assigning a new value to an index.

````python
# Example: 4) Updating List Items
fruits = ["apple", "banana", "orange"]

fruits[1] = "grape"

print(fruits)

# Press shift+enter to run the code
````

### 🗓️ Adding Items

Common ways to add items:

| Method | Meaning |
|---|---|
| `.append(x)` | Add one item to the end |
| `.insert(index, x)` | Add one item at a specific position |
| `.extend(items)` | Add multiple items from another iterable |
| `+=` | Add multiple items from another iterable |
| `*` | Copy elements of list with defined numbers of times |

````python
# Example: Adding Items
fruits = ["apple", "banana"]

fruits.append("orange")
print(fruits)
fruits.insert(1, "grape")
print(fruits)
fruits.extend(["melon", "kiwi"])
print(fruits)

mixed = ["carrot", "broccoli"]
mixed += fruits
print(mixed)
mixed += ["spinach", "lettuce"]
print(mixed)

# If you use .append() with a list, it will add the entire list as a single element. If you use .extend() or +=, it will add each element of the list individually to the existing list.
cities = ["New York", "Los Angeles", "Chicago"]
print(cities)
cities.append(["Houston", "Phoenix"])
print("After append:", cities)

sports = ["soccer", "basketball"]
print(sports * 3)

# Press shift+enter to run the code
````

### 🗓️ Removing Items

Common ways to remove items:

| Method | Meaning |
|---|---|
| `del` | Remove element with slicing |
| `.remove(x)` | Remove the first matching value |
| `.pop()` | Remove and return the last item |
| `.pop(index)` | Remove and return the item at an index |
| `.clear()` | Remove all items |

````python
# Example: Removing Items
fruits = ["apple", "banana", "orange", "banana"]

fruits.remove("banana")
last_fruit = fruits.pop()

print("List:", fruits)
print("Popped item:", last_fruit)

meals = ["breakfast", "lunch", "dinner"]
del meals[1]
print(meals)
meals.clear()
print(meals)

# Press shift+enter to run the code
````

### 🗓️ Replace elements

We can replace elements in a list using

| Method | Meaning |
|---|---|
| `[offset]` | Replace value with specific offset |
| Slicing | Replace value with slicing |


````python
# Example: Replace elements
sports = ["soccer", "basketball", "tennis", "soccer"]
print(sports)
sports[2] = "baseball"
print(sports)

countries = ["USA", "Canada", "Mexico", "Japan"]
print(countries)
countries[1:3] = ["UK", "France"]
print(countries)

# Press shift+enter to run the code
````

## 5) Sorting And Reversing Lists

Lists can be sorted or reversed.

| Tool | Meaning |
|---|---|
| `.sort()` | Sort the original list in place |
| `sorted(list)` | Return a new sorted list |
| `.reverse()` | Reverse the original list in place |
| `reversed(list)` | Return an iterator in reverse order |

⚠️ <strong>Important:</strong> `.sort()` changes the original list. `sorted()` creates a new sorted result.

````python
# Example: 5) Sorting And Reversing Lists
scores = [82, 95, 71, 88]

sorted_scores = sorted(scores)

print("Original:", scores)
print("Sorted result:", sorted_scores)

scores.sort(reverse=True)
print("Original after .sort(reverse=True):", scores)

# Press shift+enter to run the code
````

## 8) List Copying

Assignment does not copy a list. It gives another name to the same list object.

```python
a = [1, 2, 3]
b = a
```

Now `a` and `b` refer to the same list.

````python
# Example: 8) List Copying
a = [1, 2, 3]
b = a

b.append(4)

print("a:", a)
print("b:", b)
print("Same object?", a is b)

# Press shift+enter to run the code
````

There are 3 ways to copy a list.

1. `.copy()` (shallow copy) or `.deepcopy()` (deep copy) methods
2. list() function
3. slicing

````python
# Example: 8) List Copying
a = [1, 2, 3]
b = a.copy()

b.append(4)

print("a:", a)
print("b:", b)
print("Same object?", a is b)

# Shallow vs. Deep Copy
a = [[1, 2], [3, 4]]
b = a.copy()  # This creates a shallow copy of the list

b[0][0] = 99
print("a:", a)
print("b:", b)

import copy
c = copy.deepcopy(a)  # This creates a deep copy of the list

c[0][0] = 88
print("a:", a)
print("c:", c)

# Press shift+enter to run the code
````

````python
# Example: 8) List Copying
a = [1, 2, 3]
b = list(a)  # This also creates a shallow copy of the list
b.append(4)
print("a:", a)
print("b:", b)

# Press shift+enter to run the code
````

````python
# Example: 8) List Copying
a = [1, 2, 3]
b = a[:]  # This also creates a shallow copy of the list
b.append(4)
print("a:", a)
print("b:", b)

# Press shift+enter to run the code
````

### 🗓️ Nested Lists

A nested list is a list inside another list. Nested lists are common in table-like data.

````python
# Example: Nested Lists
grades = [
    ["Alice", 90],
    ["Ben", 82],
    ["Chloe", 95],
]

print(grades[0])
print(grades[0][0])
print(grades[0][1])
for i, grade in enumerate(grades):
    print(f"Student {i}: Name={grade[0]}, Score={grade[1]}")

# Press shift+enter to run the code
````

## 9) List Comprehension

A <strong>list comprehension</strong> is a compact way to create a new list from an existing iterable.

Basic syntax:

```python
new_list = [expression for item in iterable]
```

You can also add a condition:

```python
new_list = [expression for item in iterable if condition]
```

Use list comprehensions when the transformation is simple and easy to read.

````python
# Create a new list from an existing list
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]
print("Squares:", squares)

# Filter values with an if condition
even_numbers = [number for number in numbers if number % 2 == 0]
print("Even numbers:", even_numbers)

# Transform strings
names = ["alice", "bob", "chloe"]
title_names = [name.title() for name in names]
print("Title names:", title_names)

# Press shift+enter to run the code
````

## 10) Iteration With `in`, `zip()`, And `enumerate()`

Lists are iterable, so you can loop through them with `for ... in`.

Common iteration patterns:

| Pattern | Use case |
|---|---|
| `for item in list` | Read each item directly |
| `enumerate(list)` | Read each item with its index |
| `zip(list1, list2)` | Loop through two or more lists together |

⚠️ <strong>Note:</strong> `zip()` stops when the shortest iterable ends.

````python
# for ... in reads each item directly
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print("Fruit:", fruit)

print("-------------------------------")

# enumerate() gives index and value together
for index, fruit in enumerate(fruits):
    print(index, fruit)

print("-------------------------------")

# zip() combines multiple lists item by item
names = ["Alice", "Ben", "Chloe"]
scores = [90, 82, 95]

for name, score in zip(names, scores):
    print(f"{name}: {score}")

# Press shift+enter to run the code
````

## 11) Compare lists

Lists can be compared in several ways.

| Comparison | Meaning |
|---|---|
| `a == b` | Checks whether two lists have the same values in the same order |
| `a != b` | Checks whether two lists are different |
| `a is b` | Checks whether two variables refer to the same list object |
| `<`, `>`, `<=`, `>=` | Compares lists item by item from left to right |

⚠️ <strong>Important:</strong> `==` checks value equality. `is` checks object identity.

````python
# Compare list values
numbers_a = [1, 2, 3]
numbers_b = [1, 2, 3]
numbers_c = [1, 3, 2]

print("numbers_a == numbers_b:", numbers_a == numbers_b)
print("numbers_a == numbers_c:", numbers_a == numbers_c)
print("numbers_a != numbers_c:", numbers_a != numbers_c)
print("numbers_a is numbers_b:", numbers_a is numbers_b)

# Order matters when comparing lists.
print("[1, 2, 3] == [3, 2, 1]:", [1, 2, 3] == [3, 2, 1])

# Less-than and greater-than compare item by item.
print("[1, 2, 3] < [1, 2, 4]:", [1, 2, 3] < [1, 2, 4])
print("[1, 3] > [1, 2, 99]:", [1, 3] > [1, 2, 99])

# Press shift+enter to run the code
````

# 2. Tuple

## 1) What Is A Tuple?

A <strong>tuple</strong> stores multiple values in one variable, like a list, but it is immutable.

Tuples are:

| Feature | Meaning |
|---|---|
| Ordered | Items keep their position |
| Immutable | You cannot change items after creation |
| Allow duplicates | The same value can appear more than once |
| Good for fixed records | Useful for values that should stay together |

````python
# Example: 1) What Is A Tuple?
point = (10, 20)
person = ("Alice", 25, "Data Analyst")
single_item_tuple = ("Python",)

print(point)
print(person)
print(single_item_tuple)
print(type(point))
print(type(single_item_tuple))

# Press shift+enter to run the code
````

## 2) How to create a tuple

Tuples can be generated by

| Method | Meaning |
|---|---|
| `()` | Generate an empty tuple and a tuple with elements |
| `tuple()` | Convert string, tuple, dictionary & set to list|

````python
# Example: 2) How to create a tuple
empty_tuple = ()
print("Empty tuple:", empty_tuple)
list_tuple = tuple([1, 2, 3])
print("List to tuple:", list_tuple)
string_tuple = tuple("Hello")
print("String to tuple:", string_tuple)
dict_tuple = tuple({"a": 1, "b": 2})
print("Dictionary to tuple (keys):", dict_tuple)
set_tuple = tuple({10, 20, 30})
print("Set to tuple:", set_tuple)

# Press shift+enter to run the code
````

### 🗓️ Single-Item Tuple

A single-item tuple needs a comma.

| Code | Type |
|---|---|
| `("Python")` | `str` |
| `("Python",)` | `tuple` |

````python
# Example: Single-Item Tuple
not_tuple = ("Python")
real_tuple = ("Python",)

print(not_tuple, type(not_tuple))
print(real_tuple, type(real_tuple))

# Press shift+enter to run the code
````

## 2) Tuple Indexing And Unpacking

Tuples support indexing and slicing like lists.

Tuple unpacking assigns tuple items to multiple variables.

````python
# Example: 2) Tuple Indexing And Unpacking
point = (10, 20, 30)

print(point[0])
print(point[-1])
print(point[0:2])

x, y, z = point
print(x, y, z)

# Press shift+enter to run the code
````

### 🗓️ Tuple Immutability

Tuples cannot be changed in place.

```python
point = (10, 20)
point[0] = 99  # TypeError
```

If you need to change items often, use a list instead.

````python
# Example: Tuple Immutability
point = (10, 20)

# This would cause TypeError if uncommented.
# point[0] = 99

new_point = (99, point[1])
print("Original:", point)
print("New:", new_point)

# Press shift+enter to run the code
````

## 3) Tuple Methods

Tuples have fewer methods than lists because they cannot be changed.

| Method | Meaning |
|---|---|
| `.count(x)` | Count how many times a value appears |
| `.index(x)` | Return the index of the first matching value |

````python
# Example: 3) Tuple Methods
colors = ("red", "blue", "red", "green")

print(colors.count("red"))
print(colors.index("green"))

# Press shift+enter to run the code
````

## 4) Manipulate tuple

| Method | Meaning |
|---|---|
| `+` | Combine two tuples into one |
| `*` | Repeat tuple components (number is following and that means how many times) |

````python
# Example: 4) Manipulate tuple
one_tuple = (42,)
another_tuple = (56, 67)
print(one_tuple + another_tuple)

non_repeated_tuple = (1, 2, 3)
print(non_repeated_tuple * 3)

# Press shift+enter to run the code
````

## 5) Compare tuples

Tuples can be compared like lists.

| Comparison | Meaning |
|---|---|
| `a == b` | Checks whether two tuples have the same values in the same order |
| `a != b` | Checks whether two tuples are different |
| `<`, `>`, `<=`, `>=` | Compares tuples item by item from left to right |

Tuple comparison is useful when tuple values represent ordered records, such as `(year, month, day)` or `(score, name)`.

````python
# Compare tuple values
point_a = (10, 20)
point_b = (10, 20)
point_c = (10, 30)

print("point_a == point_b:", point_a == point_b)
print("point_a != point_c:", point_a != point_c)
print("point_a < point_c:", point_a < point_c)

# Tuple comparison checks items from left to right.
date_a = (2026, 6, 8)
date_b = (2026, 7, 1)

print("date_a < date_b:", date_a < date_b)

# Press shift+enter to run the code
````

## 6) Iteration With `in`, `zip()`, And `enumerate()`

Tuples are also iterable, so you can loop through them with `for ... in`.

Common iteration patterns:

| Pattern | Use case |
|---|---|
| `for item in tuple` | Read each item directly |
| `enumerate(tuple)` | Read each item with its index |
| `zip(tuple1, tuple2)` | Loop through two or more lists together |

⚠️ <strong>Note:</strong> `zip()` stops when the shortest iterable ends.

````python
# for ... in reads each item directly
fruits = ("apple", "banana", "orange")

for fruit in fruits:
    print("Fruit:", fruit)

print("-------------------------------")

# enumerate() gives index and value together
for index, fruit in enumerate(fruits):
    print(index, fruit)

print("-------------------------------")

# zip() combines multiple lists item by item
names = ("Alice", "Ben", "Chloe")
scores = (90, 82, 95)

for name, score in zip(names, scores):
    print(f"{name}: {score}")

# Press shift+enter to run the code
````

## 7) Update tuple

Tuples are immutable, so you cannot update an item in place.

```python
tuple1[0] = 99  # TypeError
```

To "update" a tuple, create a new tuple from the parts you want to keep and the new values you want to add.

⚠️ <strong>Important:</strong> Operators like `+=` do not modify the original tuple object. They create a new tuple and reassign the variable name.

````python
# Tuples cannot be updated in place.
original = (1, 2, 3)

# This would cause TypeError if uncommented.
# original[0] = 99

# Create a new tuple with the updated value.
updated = (99,) + original[1:]

print("Original:", original)
print("Updated:", updated)
print("Original id:", id(original))
print("Updated id:", id(updated))

print("-------------------------------")

# += creates a new tuple and reassigns the variable name.
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

print("Before +=:", tuple1, id(tuple1))
tuple1 += tuple2
print("After +=:", tuple1, id(tuple1))

# Press shift+enter to run the code
````

# 3. Comparison: list vs. tuple

Lists and tuples are both ordered collections, but they are used for different purposes.

| Point | List | Tuple |
|---|---|---|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Mutability | Mutable | Immutable |
| Best use | Data that may change | Fixed records or grouped values |
| Methods | Many update methods, such as `.append()` and `.pop()` | Few methods: `.count()` and `.index()` |
| Item access | Supports indexing and slicing | Supports indexing and slicing |
| Comparison | Compares values item by item | Compares values item by item |
| Memory usage | Usually uses a little more memory | Usually uses a little less memory |
| Processing speed | Flexible, but may have slightly more overhead | Can be slightly faster for fixed data |

Use a list when you expect to add, remove, sort, or update items.  
Use a tuple when the group of values should stay fixed.

⚠️ <strong>Performance note:</strong> Tuples can be a little faster and lighter because Python does not need to support in-place changes. However, the difference is usually small. Choose based on meaning first: use lists for changeable data and tuples for fixed records.

````python
# List: good when values need to change
shopping_items = ["milk", "bread"]
shopping_items.append("eggs")
print("List:", shopping_items)

# Tuple: good for fixed records
student_record = ("Alice", 25, "Data Analyst")
name, age, role = student_record
print("Tuple:", student_record)
print("Unpacked:", name, age, role)

# Both support indexing and slicing.
print("First list item:", shopping_items[0])
print("First tuple item:", student_record[0])

print("-------------------------------")

# Small performance check with timeit
# Results can differ by computer, Python version, and workload.
from timeit import timeit

list_create_time = timeit("[1, 2, 3, 4, 5]", number=1_000_000)
tuple_create_time = timeit("(1, 2, 3, 4, 5)", number=1_000_000)

list_index_time = timeit("items[2]", setup="items = [1, 2, 3, 4, 5]", number=1_000_000)
tuple_index_time = timeit("items[2]", setup="items = (1, 2, 3, 4, 5)", number=1_000_000)

list_loop_time = timeit("for item in items: item", setup="items = [1, 2, 3, 4, 5]", number=1_000_000)
tuple_loop_time = timeit("for item in items: item", setup="items = (1, 2, 3, 4, 5)", number=1_000_000)

# Creation time
print("List creation time:", list_create_time)
print("Tuple creation time:", tuple_create_time)
# Indexing time
print("List indexing time:", list_index_time)
print("Tuple indexing time:", tuple_index_time)
# Looping time
print("List loop time:", list_loop_time)
print("Tuple loop time:", tuple_loop_time)

# Press shift+enter to run the code
````

# 4. Key Terms

| Term | Meaning |
|---|---|
| Collection | A data type that can hold multiple values |
| List | An ordered, mutable collection written with square brackets `[]` |
| Tuple | An ordered, immutable collection written with parentheses `()` |
| Ordered | Items keep a position that can be accessed by index |
| Mutable | Can be changed after creation |
| Immutable | Cannot be changed after creation |
| Index | A position inside an ordered collection |
| Slicing | Extracting part of an ordered collection with `start:stop:step` |

Remember these core ideas:

- Use a list when order matters and the data needs to change.
- Use a tuple when order matters and the data should stay fixed.
- Lists are mutable; tuples are immutable.
- Assignment does not copy a mutable collection. Use `.copy()` when you need a separate shallow copy.
- `==` compares values, while `is` compares object identity.

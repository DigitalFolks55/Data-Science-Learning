# 1. Module & Import Statement

## 1) What Is A Module?

A <strong>module</strong> is a Python file that contains Python code.

A module can contain:

- variables
- functions
- classes
- executable code

When you import a module, Python loads it and makes its contents available.

Common examples of built-in or standard library modules:

| Module | Use case |
|---|---|
| `math` | Mathematical functions |
| `random` | Random values |
| `datetime` | Dates and times |
| `pathlib` | File paths |
| `json` | JSON data |
| `statistics` | Basic statistics |

````python
# Example: inspect a standard library module
import math

print(math)
print(type(math))
print(math.pi)
print(math.sqrt(16))

# Press shift+enter to run the code
````

## 2) What Does `import` Do?

The `import` statement asks Python to find, load, and bind a module name.

```python
import math
```

After this, you can access names inside the module with dot notation:

```python
math.sqrt(16)
```

This style keeps it clear where a function or value came from.

````python
# Example: use module dot notation
import statistics

scores = [80, 90, 100]

print(statistics.mean(scores))
print(statistics.median(scores))

# Press shift+enter to run the code
````

# 1-1. Import Modules

## 1) Import A Whole Module

When you import a whole module, you access its contents with `module_name.object_name`.

This is often the clearest import style for beginners.

````python
# Example: import whole modules
import random
import datetime

print(random.randint(1, 6))
print(datetime.date.today())

# Press shift+enter to run the code
````

## 2) Import Multiple Modules

You can write multiple import statements.

This is usually clearer than importing many modules on one line.

````python
# Example: import modules on separate lines
import math
import json

radius = 3
area = math.pi * radius ** 2

record = {"radius": radius, "area": area}
print(json.dumps(record, indent=2))

# Press shift+enter to run the code
````

# 1-2. Import Modules With Alias

## 1) Why Use An Alias?

An <strong>alias</strong> is a shorter or conventional name for an imported module.

Syntax:

```python
import module_name as alias
```

Aliases are common for long module names or community conventions.

| Import | Common use |
|---|---|
| `import math as m` | Short example alias |
| `import pandas as pd` | Common data-analysis convention |
| `import numpy as np` | Common numerical-computing convention |

````python
# Example: import a module with an alias
import math as m

print(m.pi)
print(m.sqrt(25))

# Press shift+enter to run the code
````

## 2) Alias Carefully

Aliases should make code easier to read.

Avoid confusing aliases such as:

```python
import math as banana
```

Use aliases that are short, familiar, or meaningful.

````python
# Example: a meaningful alias
import datetime as dt

today = dt.date.today()
print(today)

# Press shift+enter to run the code
````

# 1-3. Import Using `from`

## 1) Import Only What You Need

`from` imports specific names from a module.

Syntax:

```python
from module_name import object_name
```

After this, you use the imported name directly.

````python
# Example: import only the function we need
from math import sqrt

print(sqrt(36))

# Press shift+enter to run the code
````

## 2) Import Multiple Names

You can import multiple names from the same module.

````python
# Example: import multiple names
from math import pi, pow, sqrt

radius = 4
area = pi * pow(radius, 2)

print(area)
print(sqrt(81))

# Press shift+enter to run the code
````

## 3) `from ... import ... as ...`

You can also use aliases with `from` imports.

````python
# Example: from import with alias
from datetime import date as Date

today = Date.today()
print(today)

# Press shift+enter to run the code
````

## 4) Be Careful With `import *`

`from module import *` imports many names at once.

It is usually discouraged because it can make code harder to read and can overwrite existing names.

Prefer explicit imports.

````python
# Example: explicit import is clearer than import *
from math import ceil, floor

value = 3.7
print(floor(value))
print(ceil(value))

# Press shift+enter to run the code
````

# 2. Package

## 1) What Is A Package?

A <strong>package</strong> is a way to organize related modules in folders.

A simple package structure can look like this:

```text
my_package/
    __init__.py
    calculator.py
    formatter.py
```

- `calculator.py` is a module.
- `formatter.py` is a module.
- `my_package` is a package.
- `__init__.py` marks a regular package and can run package setup code.

Modern Python can also support namespace packages without `__init__.py`; we will cover that later.

````python
# Example: create a temporary package and import from it
import sys
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as temp_dir:
    root = Path(temp_dir)
    package_dir = root / "demo_package"
    package_dir.mkdir()

    (package_dir / "__init__.py").write_text("PACKAGE_NAME = 'demo_package'\n")
    (package_dir / "calculator.py").write_text("def add(a, b):\n    return a + b\n")

    sys.path.insert(0, str(root))
    try:
        import demo_package
        from demo_package import calculator

        print(demo_package.PACKAGE_NAME)
        print(calculator.add(2, 3))
    finally:
        sys.path.remove(str(root))
        sys.modules.pop("demo_package", None)
        sys.modules.pop("demo_package.calculator", None)

# Press shift+enter to run the code
````

## 2) Package Import Styles

You can import package modules in several ways.

| Style | Example |
|---|---|
| Import package | `import demo_package` |
| Import submodule | `import demo_package.calculator` |
| Import submodule with `from` | `from demo_package import calculator` |
| Import object from submodule | `from demo_package.calculator import add` |

````python
# Example: import an object from a package module
import sys
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as temp_dir:
    root = Path(temp_dir)
    package_dir = root / "shop"
    package_dir.mkdir()

    (package_dir / "__init__.py").write_text("")
    (package_dir / "tax.py").write_text("def add_tax(price):\n    return price * 1.1\n")

    sys.path.insert(0, str(root))
    try:
        from shop.tax import add_tax
        print(add_tax(1000))
    finally:
        sys.path.remove(str(root))
        sys.modules.pop("shop", None)
        sys.modules.pop("shop.tax", None)

# Press shift+enter to run the code
````

# 2-1. Search Module Path

## 1) How Python Finds Modules

When Python sees an import, it searches locations in `sys.path`.

`sys.path` usually includes:

- the current working directory or notebook location
- installed package locations
- standard library paths
- paths added by tools or environment settings

````python
# Example: inspect module search paths
import sys

for path in sys.path[:5]:
    print(path)

# Press shift+enter to run the code
````

## 2) Check Where A Module Came From

Many modules have a `__file__` attribute that shows where the module file is located.

Built-in modules may not have a normal file path.

````python
# Example: inspect module file locations
import math
import json

print("math file:", getattr(math, "__file__", "built-in or no file"))
print("json file:", getattr(json, "__file__", "built-in or no file"))

# Press shift+enter to run the code
````

## 3) Find A Module Spec

`importlib.util.find_spec()` can check how Python would find a module.

````python
# Example: find a module spec
import importlib.util

for module_name in ["math", "json", "not_a_real_module"]:
    spec = importlib.util.find_spec(module_name)
    print(module_name, "->", spec)

# Press shift+enter to run the code
````

# 2-2. Relative And Absolute Imports

## 1) Absolute Import

An <strong>absolute import</strong> starts from the top-level package name.

Example inside a package:

```python
from app.utils.formatter import title_text
```

Absolute imports are usually clearer because they show the full path.

## 2) Relative Import

A <strong>relative import</strong> starts from the current package location.

Examples:

| Syntax | Meaning |
|---|---|
| `from .module import name` | Import from the same package |
| `from ..module import name` | Import from the parent package |

Relative imports work inside packages, not usually from a standalone script or notebook cell.

````python
# Example: absolute and relative imports inside a temporary package
import sys
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as temp_dir:
    root = Path(temp_dir)
    app_dir = root / "app"
    app_dir.mkdir()

    (app_dir / "__init__.py").write_text("")
    (app_dir / "formatter.py").write_text("def title_text(text):\n    return text.title()\n")

    # relative import inside greeter.py: from .formatter import title_text
    (app_dir / "greeter.py").write_text(
        "from .formatter import title_text\n\n"
        "def greet(name):\n"
        "    return 'Hello, ' + title_text(name)\n"
    )

    sys.path.insert(0, str(root))
    try:
        # absolute import from outside the package
        from app.greeter import greet
        print(greet("alice"))
    finally:
        sys.path.remove(str(root))
        for module_name in ["app", "app.formatter", "app.greeter"]:
            sys.modules.pop(module_name, None)

# Press shift+enter to run the code
````

## 3) Which Should You Use?

| Import style | Good for |
|---|---|
| Absolute import | Clear, stable imports across a project |
| Relative import | Short imports between nearby modules inside the same package |

For beginner projects, prefer absolute imports unless there is a clear reason to use relative imports.

# 2-3. Namespace Packages

## 1) What Is A Namespace Package?

A <strong>namespace package</strong> is a package that can be spread across multiple folders.

It can exist without an `__init__.py` file.

This is useful when different distributions or directories contribute modules to the same package namespace.

Modern Python supports this through PEP 420.

````python
# Example: namespace package across two temporary folders
import sys
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as temp_dir:
    root = Path(temp_dir)
    part1 = root / "part1"
    part2 = root / "part2"

    (part1 / "shared_ns").mkdir(parents=True)
    (part2 / "shared_ns").mkdir(parents=True)

    # No __init__.py files: this is a namespace package example.
    (part1 / "shared_ns" / "alpha.py").write_text("VALUE = 'alpha'\n")
    (part2 / "shared_ns" / "beta.py").write_text("VALUE = 'beta'\n")

    sys.path.insert(0, str(part1))
    sys.path.insert(0, str(part2))
    try:
        import shared_ns.alpha
        import shared_ns.beta

        print(shared_ns.alpha.VALUE)
        print(shared_ns.beta.VALUE)
        print(shared_ns.__path__)
    finally:
        sys.path.remove(str(part2))
        sys.path.remove(str(part1))
        for module_name in ["shared_ns", "shared_ns.alpha", "shared_ns.beta"]:
            sys.modules.pop(module_name, None)

# Press shift+enter to run the code
````

## 2) Regular Package vs Namespace Package

| Point | Regular package | Namespace package |
|---|---|---|
| `__init__.py` | Usually exists | Usually does not exist |
| Location | Usually one folder | Can span multiple folders |
| Setup code | Can run `__init__.py` | No `__init__.py` setup code |
| Use case | Most beginner projects | Advanced packaging/distribution cases |

# 2-4. Module Or Object?

## 1) Imported Names Can Be Different Things

An imported name might refer to:

- a module
- a function
- a class
- a variable
- another object

Use `type()`, `dir()`, and `inspect` to understand what you imported.

````python
# Example: module or object?
import math
from math import sqrt
from datetime import date

print("math:", type(math))
print("sqrt:", type(sqrt))
print("date:", type(date))

print("math is module?", hasattr(math, "__file__") or hasattr(math, "__spec__"))
print("sqrt result:", sqrt(49))
print("date today:", date.today())

# Press shift+enter to run the code
````

## 2) Use `inspect` For More Detail

The `inspect` module can identify modules, functions, classes, and methods.

````python
# Example: inspect imported names
import inspect
import math
from math import sqrt
from datetime import date

objects = [math, sqrt, date, 123, "Python"]

for obj in objects:
    print(repr(obj))
    print("  is module:", inspect.ismodule(obj))
    print("  is function:", inspect.isfunction(obj) or inspect.isbuiltin(obj))
    print("  is class:", inspect.isclass(obj))

# Press shift+enter to run the code
````

## 3) Avoid Name Confusion

If you import an object directly, the module name may not exist in your namespace.

```python
from math import sqrt
```

This gives you `sqrt`, but not necessarily the name `math`.

If you need the module namespace, use:

```python
import math
```

````python
# Example: direct import gives the object name directly
from math import sqrt

print(sqrt(64))

# In a fresh script, math would not be available unless you also write: import math
# This notebook may already have math from previous cells, so this is a conceptual warning.

# Press shift+enter to run the code
````

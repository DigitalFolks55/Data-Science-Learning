# 1. Dictionary

## 1) What Is A Dictionary?

A <strong>dictionary</strong> stores data as key-value pairs.

```python
student = {"name": "Alice", "score": 90}
```

In this example:

| Part | Meaning |
|---|---|
| `"name"` | key |
| `"Alice"` | value |
| `"score"` | key |
| `90` | value |

Dictionaries are useful when each value has a clear label.

````python
# What is a dictionary in Python?
student = {
    "name": "Alice",
    "age": 25,
    "score": 90,
    "passed": True,
}

print("student:", student)
print("Type:", type(student))

# Press shift+enter to run the code
````

### 🗓️ Dictionary Features

Dictionaries are:

| Feature | Meaning |
|---|---|
| Key-value based | Values are accessed by keys, not numeric positions |
| Mutable | You can add, update, and remove key-value pairs |
| Unique keys | A key can appear only once in the same dictionary |
| Ordered by insertion | Modern Python keeps insertion order |

⚠️ <strong>Important:</strong> Dictionary keys must be hashable. Common key types include strings, numbers, booleans, and tuples containing hashable values.

````python
student = {
    "name": "Alice",
    "name": "Alicia",
    "score": 90,
}

print("student:", student)
print("The second value for the same key replaced the first one.")

# Press shift+enter to run the code
````

### 🗓️ Dictionary Is An Object

A dictionary object has:

| Property | Example |
|---|---|
| Value | `{"name": "Alice", "score": 90}` |
| Type | `dict` |
| Identity | result of `id(student)` |
| Methods | `.get()`, `.keys()`, `.values()`, `.items()`, and many more |

````python
# What are the properties of a dictionary in Python?
student = {"name": "Alice", "score": 90}

print("Value:", student)
print("Type:", type(student))
print("Identity:", id(student))
print("Length:", len(student))
print("Available methods include:")
print([name for name in dir(student) if not name.startswith("_")])

# Press shift+enter to run the code
````

## 2) Creating Dictionaries

You can create dictionaries in several ways.

| Syntax | Example |
|---|---|
| Literal | `{"name": "Alice", "score": 90}` |
| Empty dictionary | `{}` |
| `dict()` | `dict(name="Alice", score=90)` |
| From pairs | `dict([("name", "Alice"), ("score", 90)])` |

````python
# Create dictionaries using different methods
student1 = {"name": "Alice", "score": 90}
student2 = {}
student3 = dict(name="Ben", score=82)
student4 = dict([("name", "Chloe"), ("score", 95)])
student5 = dict((["name", "David"], ["score", 88]))

twocharactorlist = dict(["ab", "cd"])
twocharactortuple = dict(("ef", "gh"))

print("student1:", student1)
print("student2:", student2)
print("student3:", student3)
print("student4:", student4)
print("student5:", student5)

print("twocharactorlist:", twocharactorlist)
print("twocharactortuple:", twocharactortuple)

# Press shift+enter to run the code
````

### 🗓️ Dictionary Versus List

Use a dictionary when labels matter.

| Data | Better structure | Why |
|---|---|---|
| Scores in order | List | Position matters |
| One student's profile | Dictionary | Each value needs a label |
| Many student profiles | List of dictionaries | Multiple records with labels |

````python
# List: position-based
student_list = ["Alice", 25, 90]

# Dictionary: label-based
student_dict = {"name": "Alice", "age": 25, "score": 90}

print("student_list[0]:", student_list[0])
print("student_dict['name']:", student_dict["name"])

# Press shift+enter to run the code
````

## 3) Accessing Values

Use square brackets with a key to access a value.

```python
student["name"]
```

If the key does not exist, square brackets raise `KeyError`.

````python
# Accessing dictionary values using keys
student = {"name": "Alice", "age": 25, "score": 90}

print("student['name']:", student["name"])
print("student['score']:", student["score"])

# This would cause KeyError if uncommented.
# print(student["email"])

# Press shift+enter to run the code
````

### 🗓️ Safe Access With `.get()`

`.get()` returns `None` when a key does not exist.  
You can also provide a default value.

````python
# Using the get() method to access dictionary values
student = {"name": "Alice", "age": 25, "score": 90}

print("student.get('name'):", student.get("name"))
print("student.get('email'):", student.get("email"))
print("student.get('email', 'No email registered'):", student.get("email", "No email registered"))

# Press shift+enter to run the code
````

### 🗓️ Membership Checks

For dictionaries, `in` checks keys, not values.

````python
# Using the in operator to check for key existence in a dictionary
student = {"name": "Alice", "age": 25, "score": 90}

print("name in student:", "name" in student)
print("Alice in student:", "Alice" in student)
print("Alice in student.values():", "Alice" in student.values())

# Press shift+enter to run the code
````

## 4) Adding And Updating Values

Dictionaries are mutable.

You can add a new key or update an existing key using assignment.

````python
# Adding and updating dictionary entries
student = {"name": "Alice", "score": 90}

student["age"] = 25
student["score"] = 95

print("Updated student:", student)

# Press shift+enter to run the code
````

### 🗓️ Updating Multiple Values

Use `.update()` to add or update multiple key-value pairs.

````python
# Using the update() method to add or update dictionary entries
student = {"name": "Alice", "score": 90}

student.update({"age": 25, "score": 95, "passed": True})

print("Updated student:", student)

# Press shift+enter to run the code
````

### 🗓️ Merging Multiple dictionary

Use `.update()` or `{**dict1, **dict2}` to add or update multiple key-value pairs.

````python
# Merging dictionaries using the update() method and dictionary unpacking
dict1 = {"name": "Alice", "score": 90}
dict2 = {"age": 25, "score": 95, "passed": True}

dict1.update(dict2)

print("Merged dictionary (using update()):", dict1)

dict1 = {"name": "Alice", "score": 90}
dict2 = {"school": "ABC High School", "score": 95, "passed": True}

dict4 = {**dict1, **dict2}

print("Merged dictionary (using dictionary unpacking):", dict4)

# Press shift+enter to run the code
````

### 🗓️ Removing Values

Common ways to remove items:

| Method | Meaning |
|---|---|
| `.pop(key)` | Remove and return the value for a key |
| `.pop(key, default)` | Return default if key does not exist |
| `del dictionary[key]` | Delete a key-value pair |
| `.clear()` | Remove all key-value pairs |

````python
# Removing dictionary entries using pop() and del
student = {"name": "Alice", "age": 25, "score": 90}

removed_score = student.pop("score")
student.pop("email", "No email to remove")

del student["age"]

print("Removed score:", removed_score)
print("Updated student:", student)

# Press shift+enter to run the code
````

## 5) Dictionary Views

Dictionaries provide views of keys, values, and items.

| Method | Meaning |
|---|---|
| `.keys()` | Get dictionary keys |
| `.values()` | Get dictionary values |
| `.items()` | Get key-value pairs |

These are useful when looping through dictionary data.

### 🗓️ Looping Through A Dictionary

The most common dictionary loop uses `.items()`.

````python
# Iterating through dictionary keys, values, and items
student = {"name": "Alice", "age": 25, "score": 90}

print("Iterating through items:")
for key, value in student.items():
    print(key, "->", value)

# Press shift+enter to run the code
````

````python
# Iterating through dictionary keys, values, and items
scores = {"Alice": 90, "Ben": 82, "Chloe": 95}

print("Iterating through scores:")
for name, score in scores.items():
    print(f"{name}: {score}")

# Press shift+enter to run the code
````

## 6) Nested Dictionaries

A nested dictionary is a dictionary inside another dictionary.

Nested dictionaries are useful for structured data.

````python
# Nested dictionaries and accessing nested values
students = {
    "S001": {"name": "Alice", "score": 90},
    "S002": {"name": "Ben", "score": 82},
    "S003": {"name": "Chloe", "score": 95},
}

print("Student S001:", students["S001"])
print("Name of student S001:", students["S001"]["name"])
print("Score of student S003:", students["S003"]["score"])

# Press shift+enter to run the code
````

### 🗓️ Looping Through Nested Dictionaries

Use nested access when each value is another dictionary.

````python
# Iterating through nested dictionaries
students = {
    "S001": {"name": "Alice", "score": 90},
    "S002": {"name": "Ben", "score": 82},
    "S003": {"name": "Chloe", "score": 95},
}

print("Iterating through students:")
for student_id, profile in students.items():
    print(student_id, profile["name"], profile["score"])

# Press shift+enter to run the code
````

## 7) Dictionary Copying

Assignment does not copy a dictionary. It gives another name to the same dictionary object.

````python
# Understanding dictionary references and copies
student = {"name": "Alice", "score": 90}
alias = student

alias["score"] = 95

print("student:", student)
print("alias:", alias)
print("Same object?", student is alias)

# Press shift+enter to run the code
````

````python
# Understanding dictionary references and copies
student = {"name": "Alice", "score": 90}
copy_student = student.copy()

copy_student["score"] = 95

print("student:", student)
print("copy_student:", copy_student)
print("Same object?", student is copy_student)

# Press shift+enter to run the code
````

### 🗓️ Shallow Copy Caution

`.copy()` creates a shallow copy. If a dictionary contains nested mutable objects, the nested objects are still shared.

`.deepcopy()` copy a dictionary as a new object. It is necessary for nested dictionaries to copy dictionaries and to update values.

````python
# Shallow copy of dictionaries
student = {"name": "Alice", "scores": [80, 90]}
copy_student = student.copy()

copy_student["scores"].append(100)

print("student:", student)
print("copy_student:", copy_student)

# Press shift+enter to run the code
````

````python
# Deep copy of dictionaries
import copy

student = {"name": "Alice", "scores": [80, 90]}
student_copy = copy.deepcopy(student)
student_copy["scores"][1] = 95

print("student:", student)
print("student_copy:", student_copy)

# Press shift+enter to run the code
````

## 8) Compare dictionaries

Lists can be compared in several ways.

| Comparison | Meaning |
|---|---|
| `a == b` | Checks whether two dictionaries have the same values in the same order |
| `a != b` | Checks whether two dictionaries are different |
| `a is b` | Checks whether two variables refer to the same dictionariy object |
| `<`, `>`, `<=`, `>=` | ⚠️ <strong>Not supported</strong> |


````python
# Comparing dictionaries using equality and identity operators
a = {1: 1, 2: 2, 3: 3}
b = {3: 3, 1: 1, 2: 2}
c = {1: 1, 2: 2, 3: 4}

print("a == b:", a == b)
print("a == c:", a == c)

print("a is b:", a is b)
print("a is c:", a is c)

print("a != b:", a != b)
print("a != c:", a != c)

# This will raise a TypeError because dictionaries do not support forward comparison operators like <, >, <=, >=.
# print(a < b)
# print(a > c)

# Press shift+enter to run the code
````

### 🗓️ Counting Values

Use a dictionary when you want to count how many times each value appears.

````python
# Counting occurrences of words using a dictionary
words = ["python", "sql", "python", "excel", "sql", "python"]
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print("Word counts:", counts)

# Press shift+enter to run the code
````

### 🗓️ Grouping Values

Use a dictionary when one key should collect multiple related values.

````python
# Grouping records by a specific key using a dictionary
records = [
    {"department": "Sales", "name": "Alice"},
    {"department": "Data", "name": "Ben"},
    {"department": "Sales", "name": "Chloe"},
]

groups = {}

for record in records:
    department = record["department"]
    groups.setdefault(department, []).append(record["name"])

print("Groups:", groups)

# Press shift+enter to run the code
````

## 10) Iteration with for & in

Dictionaries are iterable and there are 3 ways to do.

1) Iterable with keys
2) Iterable with values
3) Iterable with keys and values

````python
# Iterating over the keys of the dictionary
capitals_dict = {"USA": "Washington, D.C.", "France": "Paris", "Japan": "Tokyo"}

for country in capitals_dict:
    print(f"Country: {country}")

print("-"*20)

for country in capitals_dict.keys():
    print(f"Country: {country}")

# Press shift+enter to run the code
````

## 11) Dictionary comprehension

A <strong>dictionary comprehension</strong> is a compact way to create a new dictionary from an existing iterable.

Basic syntax:

```python
new_dict = {key_expression: value_expression for expression in iterable}
```

You can also add a condition:

```python
new_dict = {key_expression: value_expression for expression in iterable if condition}
```

Use dictionary comprehensions when the transformation is simple and easy to read.

````python
# Dictionary comprehension to count occurrences of elements in a list
country = ["USA", "Canada", "Mexico", "USA", "Canada", "USA"]
count_country = {c: country.count(c) for c in set(country)}
print(count_country)

# Press shift+enter to run the code
````

````python
# Compare the performance of counting letters in a word using dictionary comprehension with and without set()
from timeit import timeit

word = "banana"
letter_count1 = timeit(lambda: {letter: word.count(letter) for letter in word}, number=1000)
letter_count2 = timeit(lambda: {letter: word.count(letter) for letter in set(word)}, number=1000)  # Faster! a counts several times without set()

print(f"Letter count 1: {letter_count1} seconds")
print({letter: word.count(letter) for letter in word})
print(f"Letter count 2: {letter_count2} seconds")
print({letter: word.count(letter) for letter in set(word)})

# Press shift+enter to run the code
````

````python
# Dictionary comprehension to count occurrences of specific letters in a word
word = "banana"
check = "ab"

letter_count = {letter: word.count(letter) for letter in set(word) if letter in check}
print(letter_count)

# Press shift+enter to run the code
````

## 1) What Is A Set?

A <strong>set</strong> stores unique values.

Sets are:

| Feature | Meaning |
|---|---|
| Unordered | Items do not keep a fixed position |
| Mutable | You can add and remove items |
| Unique | Duplicate values are removed |
| Fast for membership checks | Good for checking whether something exists |

Sets are useful when you care about uniqueness, not order.

````python
# What is a set in Python?
numbers = {1, 2, 2, 3, 3, 3}
skills = {"Python", "SQL", "Python"}

print("numbers:", numbers)
print("skills:", skills)
print("type of skills:", type(skills))

# Press shift+enter to run the code
````

## 2) Adding, Removing, And Checking Set Items

Common set methods:

| Method | Meaning |
|---|---|
| `.add(x)` | Add one item |
| `.update(items)` | Add multiple items |
| `.remove(x)` | Remove item; error if missing |
| `.discard(x)` | Remove item; no error if missing |
| `.clear()` | Remove all items |

````python
# Adding and removing elements from a set
skills = {"Python", "SQL"}

skills.add("Excel")
skills.update(["Pandas", "Python"])
skills.discard("Tableau")

print("skills:", skills)
print("Python in skills:", "Python" in skills)
print("Java in skills:", "Java" in skills)

# Press shift+enter to run the code
````

### 🗓️ `remove()` Versus `discard()`

`remove()` raises an error when the item does not exist.  
`discard()` does nothing when the item does not exist.

Use `discard()` when missing values are acceptable.

````python
# Removing elements from a set using discard() and remove()
skills = {"Python", "SQL"}

skills.discard("Java")
print("Updated skills:", skills)

# This would cause KeyError if uncommented.
# skills.remove("Java")

# Press shift+enter to run the code
````

## 3) Set Operations

Sets are useful for comparing groups.

| Operation | Python syntax | Meaning |
|---|---|---|
| Union | `a | b` or `a.union(b)` | Items in either set |
| Intersection | `a & b` or `a.intersection(b)` | Items in both sets |
| Difference | `a - b` or `a.difference(b)` | Items in `a` but not `b` |
| Symmetric difference | `a ^ b` | Items in either set, but not both |
| Subset | `a <= b` or `a.issubset(b)` | `a` is subset of `b` |
| Proper subset | `a < b` | `a` is a proper subset of `b` |
| Superset | `a >= b` or `a.issuperset(b)` | `a` is supersubset of `b` |
| Proper superset | `a > b` | `a` is a proper superset of `b` |

````python
# Performing set operations: union, intersection, difference, and symmetric difference
python_students = {"Alice", "Ben", "Chloe"}
sql_students = {"Ben", "David", "Chloe"}

print("Union:", python_students | sql_students)
print("Intersection:", python_students & sql_students)
print("Python only:", python_students - sql_students)
print("One course only:", python_students ^ sql_students)
print("Subset:", python_students <= sql_students)
print("Proper subset:", python_students < sql_students)
print("Superset:", python_students >= sql_students)
print("Proper superset:", python_students > sql_students)

# Press shift+enter to run the code
````

⚠️ <strong>Important:</strong> A set does not preserve the original order as a learning rule. If order matters, use another approach later.


You can also add a condition:

```python
new_set = {expression: expression for expression in iterable if condition}
```

Use set comprehensions when the transformation is simple and easy to read.

## 7) Set Copying

Assignment does not copy a set. It gives another name to the same set object.

````python
# Understanding set references and copies
skills = {"Python", "SQL", "Excel"}
alias = skills

alias.add("Tableau")

print("skills:", skills)
print("alias:", alias)
print("Same object?", skills is alias)

# Press shift+enter to run the code
````

````python
# Understanding set references and copies
skills = {"Python", "SQL", "Excel"}
copy_skills = skills.copy()

copy_skills.add("Tableau")

print("skills:", skills)
print("copy_skills:", copy_skills)
print("Same object?", skills is copy_skills)

# Press shift+enter to run the code
````

### 🗓️ Shallow Copy And Deep Copy

`.copy()` creates a shallow copy of a set.

For most beginner set examples, the values are immutable objects such as strings, numbers, or tuples. In those cases, `.copy()` is usually enough.

`copy.deepcopy()` can also copy a set as a new object. For a simple set, it looks similar to `.copy()`.

````python
# Deep copy of sets
import copy

skills = {"Python", "SQL", "Excel"}
deepcopy_skills = copy.deepcopy(skills)

deepcopy_skills.add("Tableau")

print("skills:", skills)
print("deepcopy_skills:", deepcopy_skills)
print("Same object?", skills is deepcopy_skills)

# Press shift+enter to run the code
````

⚠️ <strong>Important:</strong> A set cannot contain unhashable mutable objects such as lists, dictionaries, or other sets. If you need an immutable set inside another set, use `frozenset()`.

# 4. Key Terms

| Term | Meaning |
|---|---|
| Dictionary | A mutable `dict` collection that stores key-value pairs |
| Key | A label used to access a value in a dictionary |
| Value | The data stored under a key |
| Key-value pair | One key and its related value |
| Hashable | Safe to use as a dictionary key; examples include strings, numbers, booleans, and suitable tuples |
| `KeyError` | Error raised when square brackets access a missing dictionary key |
| `.get()` | Safely read a value and optionally provide a default |
| `.update()` | Add or update multiple key-value pairs |
| `.pop()` | Remove a key-value pair and return the removed value |
| Dictionary view | A live view returned by `.keys()`, `.values()`, or `.items()` |
| `.items()` | Return key-value pairs, commonly used when looping through a dictionary |
| Nested dictionary | A dictionary stored inside another dictionary |
| Shallow copy | A copy where nested mutable objects can still be shared |
| Deep copy | A copy that also copies nested objects into new objects |
| Dictionary comprehension | Compact syntax for creating a dictionary from an iterable |
| | |
| Set | A mutable `set` collection that stores unique values |
| Empty set | Created with `set()` because `{}` creates an empty dictionary |
| Set copy | A new set object created with `.copy()` or `copy.deepcopy()` |
| Membership check | Checking whether a key or value exists with `in` |
| `.add()` | Add one item to a set |
| `.remove()` | Remove a set item and raise an error if it is missing |
| `.discard()` | Remove a set item without raising an error if it is missing |
| Union | Set operation that combines items from both sets |
| Intersection | Set operation that keeps items found in both sets |
| Difference | Set operation that keeps items in one set but not another |
| Symmetric difference | Set operation that keeps items that are in either set, but not both |
| Subset | Set comparison that checks whether every item is included in another set |
| Superset | Set comparison that checks whether a set contains every item from another set |
| Set comprehension | Compact syntax for creating a set from an iterable |
| `frozenset` | An immutable set created with `frozenset()` |

Remember these core ideas:

- Use a dictionary when each value needs a meaningful label.
- Dictionary lookup uses keys, not numeric positions.
- Use `.get()` when a key may be missing.
- Be careful with copying dictionaries that contain nested mutable objects.
- Use comprehensions when a dictionary or set transformation is simple and readable.
- Use a set when uniqueness and membership checks matter more than order.
- Use `.copy()` when you need to update a copied set without changing the original set.
- Use set operations to compare groups, and use `frozenset()` when you need an immutable set.

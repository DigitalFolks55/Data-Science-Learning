# 1. Function Basics

## 1) What Is A Function?

A <strong>function</strong> is a reusable block of code that performs a specific task.

Functions are useful when you want to:

| Use case | Meaning |
|---|---|
| Reuse code | Write logic once and call it many times |
| Organize code | Give a clear name to a task |
| Accept input | Use parameters to receive values |
| Produce output | Use `return` to send a result back |

Python already has built-in functions such as `print()`, `len()`, `type()`, and `sum()`. You can also define your own functions.

````python
# Example: 1) What Is A Function?
values = [10, 20, 30]

print(len(values))  # Python built-in function to get the length of a list
print(sum(values))  # Python built-in function to get the sum of a list
print(type(values))  # Python built-in function to get the type of a variable

# Press shift+enter to run the code
````

## 2) Function Definition With `def`

Use `def` to define a function. The function naming convension follows the same rules as those for variables: Refer to "Python fundamental - Everything in Python is an Object.ipynb".

Basic syntax:

```python
def function_name():
    code_to_run
```

The indented block belongs to the function. If you want not to do anything with this function, then you can put pass.

````python
# Example: 2) Function Definition With def
def say_hello():
    print("Hello, Python!")

say_hello()

# Press shift+enter to run the code
````

## 3) Call Function By `function()`

Defining a function does not run it immediately. To run a function, call it with parentheses.

```python
function_name()
```

Parentheses tell Python to execute the function.

````python
# Example: 3) Call Function By function()
def show_course():
    print("Python Fundamentals")

print("Before function call")
show_course()
print("After function call")

# Press shift+enter to run the code
````

## 4) Parameters And Arguments

A <strong>parameter</strong> is a variable name in the function definition.

An <strong>argument</strong> is the actual value passed when the function is called.

| Word | Where it appears | Example |
|---|---|---|
| Parameter | Function definition | `def greet(name):` |
| Argument | Function call | `greet("Alice")` |

````python
# Example: 4) Parameters And Arguments
def greet(name):  # name is a parameter
    print("Hello,", name)

greet("Alice")  # "Alice" is an argument
greet("Ben")

# Press shift+enter to run the code
````

## 5) Return Values

`return` sends a result back to the caller.

A function can print a value, return a value, or both. In data work, returning a value is usually more reusable.

````python
# Example: 5) Return Values
def add(a, b):
    return a + b

result = add(10, 5)
print("result:", result)
print("result * 2:", result * 2)

# Press shift+enter to run the code
````

## 6) Docstring

A <strong>docstring</strong> explains what a function does.

Place it as the first statement inside the function body. You can inspect it with `help()` or `function.__doc__`.

````python
# Example: 6) Docstring
def calculate_area(width, height):
    """Return the area of a rectangle."""
    return width * height

print("Area:", calculate_area(4, 5))
print("Docstring:", calculate_area.__doc__)

# Press shift+enter to run the code
````

# 2. Function Arguments

## 1) Positional And Keyword Arguments

A <strong>positional argument</strong> is matched by order.

A <strong>keyword argument</strong> is matched by parameter name.

| Call style | Example | Meaning |
|---|---|---|
| Positional | `introduce("Alice", "Data")` | Values are matched by position |
| Keyword | `introduce(role="Data", name="Alice")` | Values are matched by name |

````python
# Example: 1) Positional And Keyword Arguments
def introduce(name, role):
    print(f"{name} works in {role}.")

introduce("Alice", "Data")
introduce(role="Data", name="Ben")

# Press shift+enter to run the code
````

## 2) Default Arguments

A default argument gives a parameter a value when the caller does not provide one.

```python
def greet(name="Guest"):
    ...
```

⚠️ <strong>Important:</strong> Parameters with default values come after parameters without default values.

````python
# Example: 2) Default Arguments
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Alice")
greet("Ben", "Good morning")

# Press shift+enter to run the code
````

````python
def default_list(arg, result=[]):
    print("before result", result)
    result.append(arg)
    print("after result", result)

default_list(1)  # function is called with a default list
default_list(2)  # function is called again, but result contains the previous value of list

# Press shift+enter to run the code
````

````python
def default_list_fixed(arg, result=None):
    if result is None:
        result = []
    print("before result", result)
    result.append(arg)
    print("after result", result)

default_list_fixed(1)  # function is called with a default list
default_list_fixed(2)  # function is called again, but result is a new list

# Press shift+enter to run the code
````

````python
def default_none(arg, result=None):  # None is immutable, so it is safe to use as a default value
    print(result)
    if result is None:
        result = []
    print("before result", result)
    result.append(arg)
    print("after result", result)

default_none(1)  # function is called with a default list
default_none(2)  # function is called again, but result is a new list

# Press shift+enter to run the code
````

## 3) Using `*args`

`*args` collects extra positional arguments into a tuple.

Use it when the number of positional values can change.

````python
# Example: 3) Using *args
def total_scores(*scores):
    print("scores:", scores)
    return sum(scores)

print(total_scores(80, 90))
print(total_scores(70, 85, 95))

# Press shift+enter to run the code
````

## 4) Using `**kwargs`

`**kwargs` collects extra keyword arguments into a dictionary.

Use it when named options can change.

````python
# Example: 4) Using **kwargs
def show_profile(**profile):
    print(profile)
    for key, value in profile.items():
        print(key, "->", value)

show_profile(name="Alice", role="Data Analyst", city="Tokyo")

# Press shift+enter to run the code
````

## 5) Unpacking Arguments With `*` And `**`

You can also use `*` and `**` when calling a function.

| Syntax | Meaning |
|---|---|
| `*list_or_tuple` | Unpack positional arguments |
| `**dictionary` | Unpack keyword arguments |

````python
# Example: 5) Unpacking Arguments With * And **
def describe_student(name, score, course):
    print(f"{name} scored {score} in {course}.")

student_values = ["Alice", 92, "Python"]
describe_student(*student_values)

student_info = {"name": "Ben", "score": 85, "course": "SQL"}
describe_student(**student_info)

# Press shift+enter to run the code
````

````python
# Example: 5) Unpacking And Packing Arguments
def unpacking_packing(*args, **kwargs):
    print("args:", args, "unpacked_args:", *args)
    print("kwargs:", kwargs)
    # print("unpacked_kwargs:", **kwargs)  # This line will raise a SyntaxError because you cannot unpack keyword arguments directly in a print statement.

unpacking_packing(1, 2, 3, name="Alice", role="Data Analyst")

# Press shift+enter to run the code
````

## 6) Keyword-Only Arguments

A <strong>keyword-only argument</strong> must be passed by name.

Place `*` before keyword-only parameters.

```python
def function_name(value, *, option=True):
    ...
```

This makes function calls clearer when options are important.

````python
# Example: 6) Keyword-Only Arguments
def format_price(amount, *, currency="JPY"):
    return f"{amount} {currency}"

print(format_price(1200, currency="JPY"))
print(format_price(25, currency="USD"))
# print(format_price(1200, "JPY"))  # This would raise TypeError

# Press shift+enter to run the code
````

## 7) Mutable And Immutable Arguments

Function arguments are object references.

| Type of object | Example | Common behavior inside a function |
|---|---|---|
| Immutable | `int`, `float`, `str`, `tuple` | Reassignment creates a new object |
| Mutable | `list`, `dict`, `set` | In-place changes can affect the original object |

Be careful when a function changes a mutable object.

````python
# Example: 7) Mutable And Immutable Arguments
def add_bonus(score):
    score += 10
    print("inside immutable example:", score)

def add_item(items):
    items.append("new")
    print("inside mutable example:", items)

score = 80
tags = ["python"]

add_bonus(score)
print("outside immutable example:", score)

add_item(tags)
print("outside mutable example:", tags)

# Press shift+enter to run the code
````

### 🗓️ Avoid Mutable Default Arguments

A mutable default argument can keep changes between function calls.

Use `None` as the default, then create the mutable object inside the function.

````python
# Example: Avoid Mutable Default Arguments
def add_tag(tag, tags=None):
    if tags is None:
        tags = []
    tags.append(tag)
    return tags

print(add_tag("python"))
print(add_tag("sql"))

# Press shift+enter to run the code
````

# 3. Function Patterns

## 1) Functions As First-Class Objects

In Python, functions are <strong>first-class objects</strong>.

This means you can:

| Action | Example |
|---|---|
| Assign a function to a variable | `operation = add` |
| Pass a function as an argument | `apply(add, 2, 3)` |
| Return a function from another function | `return inner_function` |

````python
# Example: 1) Functions As First-Class Objects
def double(number):
    return number * 2

operation = double
print(operation(5))

def apply_function(function, value):
    return function(value)

print(apply_function(double, 10))

# Press shift+enter to run the code
````

## 2) Function In Function

You can define a function inside another function.

The inner function is useful when it belongs only to the outer function's task.

````python
# Example: 2) Function In Function
def clean_and_title(text):
    def clean(value):
        return value.strip().lower()

    cleaned = clean(text)
    return cleaned.title()

print(clean_and_title("  python fundamentals  "))

# Press shift+enter to run the code
````

## 3) Closure

A <strong>closure</strong> is a function that remembers variables from the outer function where it was created, even after that outer function has finished running.  

````python
# Example: 3) Closure
def outer_function(word):
    def inner_function():
        print(f"Attention: {word}!")
    return inner_function

a = outer_function("Python")
b = outer_function("Data Science")

print("Check type")
print("type(a):", type(a))  # <class 'function'>
print("type(b):", type(b))  # <class 'function'>
print("-"*8)

print("a:", a)  # It is a function object and a closure
print("b:", b)  # It is a function object and a closure
print("-"*8)

print("Check whether the closures remember the value of 'word'")
a()  # Attention: Python!
b() # Attention: Data Science!

print("closure value in a:", a.__closure__[0].cell_contents)  # Python
print("closure value in b:", b.__closure__[0].cell_contents)  # Data Science

# Press shift+enter to run the code
````

````python
# Example: 3) Closure
def make_multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(10))  # 20
print(triple(10))  # 30

# Press shift+enter to run the code
````

## 4) Lambda

A <strong>lambda function</strong> is a small anonymous function.

Basic syntax:

```python
lambda parameter: expression
```

Use lambda for short, simple functions. Use `def` when logic needs a name or multiple steps.  
lambda is useful to define call back functions in GUI.

````python
# Example: 4) Lambda
numbers = [1, 2, 3, 4]
squared = list(map(lambda number: number ** 2, numbers))

students = [("Alice", 90), ("Ben", 82), ("Chloe", 95)]
sorted_students = sorted(students, key=lambda student: student[1])

print(squared)
print(sorted_students)

# Press shift+enter to run the code
````

## 5) Generator

A <strong>generator function</strong> is an object to create sequence.   
It uses `yield` to produce values one at a time.

Generators are useful when you want to process a sequence without building the whole list at once.

````python
# Example: 5) Generator
def countdown(start):
    while start > 0:
        yield start
        start -= 1

generator1 = countdown
generator2 = countdown(3)

print("Check type")
print("type of generator1:", type(generator1))  # <class 'function'>
print("type of generator2:", type(generator2))  # <class 'generator'>
print("-"*8)

print("Iterating over generator2:")
for number in generator2:
    print(number)
print("-"*8)

# you want to use this generator again, you need to create a new generator object
print("Iterating over generator2 again:")
for number in generator2:
    print(number)

# Press shift+enter to run the code
````

### 🗓️  Generator comprehension

A generator comprehension (more commonly called a generator expression) is a concise way to create a generator.   
Unlike a list comprehension, it does not compute all values immediately.   
Instead, it produces values one at a time as they are requested.

````python
# Generator Comprehension
generator_comprehension = (x ** 2 for x in [1,2,3,4,5])

print("type of generator_comprehension:", type(generator_comprehension))  # <class 'generator'>
print("-"*8)
for number in generator_comprehension:
    print(number)

# Press shift+enter to run the code
````

## 6) Decorator

A <strong>decorator</strong> is a function that wraps another function to add behavior.

Decorator syntax:

```python
@decorator_name
def function_name():
    ...
```

Decorators contain
1) `*args` and `**kwargs`
2) Function in fuction
3) Function as argument

Decorators are common for logging, timing, validation, permissions, and web frameworks.

````python
# Example: 6) Decorator
def announce(function):
    def wrapper_announce(*args, **kwargs):
        print("Running function:", function.__name__)
        print("Start")
        result = function(*args, **kwargs)
        print("End")
        return result
    return wrapper_announce

def another_announce(function):
    def wrapper_another_announce(*args, **kwargs):
        print("Running function:", function.__name__)
        print("Before function call")
        result = function(*args, **kwargs)
        print("After function call")
        return result
    return wrapper_another_announce

@announce
def greet(name):
    print("Hello,", name)

print("Run greet function with @announce decorator:")
greet("Alice")
print("-"*8)

@announce
@another_announce  # execute another_announce first, then announce
def greet(name):
    print("Hello,", name)

print("Run greet function with @announce and @another_announce decorators:")
greet("Alice")
print("-"*8)

@another_announce
@announce  # execute announce first, then another_announce
def greet(name):
    print("Hello,", name)

print("Run greet function with @another_announce and @announce decorators:")
greet("Alice")

# Press shift+enter to run the code
````

## 7) Recursion

A <strong>recursive function</strong> calls itself.

Every recursive function needs:

| Part | Meaning |
|---|---|
| Base case | A condition that stops recursion |
| Recursive case | The function calls itself with a smaller problem |

Use recursion carefully because too many calls can raise `RecursionError`.

````python
# Example: 7) Recursion
def factorial(number):
    if number == 0:
        return 1
    return number * factorial(number - 1)

print(factorial(5))

# Press shift+enter to run the code
````

# 4. Namespace And Scope

## 1) Namespace And Scope

A <strong>namespace</strong> is a place where names are stored.

A <strong>scope</strong> is the area where a name can be used.

Common scopes follow the LEGB rule:

| Scope | Meaning |
|---|---|
| Local | Names inside the current function |
| Enclosing | Names in an outer function |
| Global | Names at the top level of a file or notebook |
| Built-in | Names provided by Python, such as `len` and `print` |

````python
# Example: 1) Namespace And Scope
course = "Python"  # global name

def show_scope():
    topic = "functions"  # local name
    print("course:", course)
    print("topic:", topic)
    print("local variables:", locals())

show_scope()

# Check gloabl variables; modified to only show the variable named "course" which we defined above
print("global variables:", [{global_variable[0]: global_variable[1]} for global_variable in globals().items() if global_variable[0] == "course"])

# Press shift+enter to run the code
````

## 2) `global` And `nonlocal`

`global` tells Python to use a global name.

`nonlocal` tells Python to use a name from an enclosing function.

Use these carefully. Returning values is often clearer.

````python
# Example: 2) global And nonlocal
counter = 0

def increase_global_counter():
    global counter
    counter += 1

increase_global_counter()
print("global counter:", counter)

def make_counter():
    count = 0

    def increase():
        nonlocal count
        count += 1
        return count

    global counter

    return increase, counter

counter_function, global_counter = make_counter()
print("nonlocal:", counter_function(), "global:", global_counter)
print("nonlocal:", counter_function(), "global:", global_counter)

# Press shift+enter to run the code
````

## 3) `_` And `__` In Function & Variable Names

Python uses underscores in names as conventions and special mechanisms.

| Pattern | Meaning |
|---|---|
| `_helper()` | Internal-use helper by convention |
| `__name` | Name mangling inside classes |
| `__call__()` | Special method name, also called a dunder method |
| `_` | Temporary or unused value by convention |

For regular functions, a single leading underscore usually means "internal helper." Double-underscore names are most important inside classes.

The same underscore patterns can also appear in variable names.

| Variable Pattern | Meaning |
|---|---|
| `_value` | Internal-use variable by convention |
| `value_` | Used when `value` would conflict with a Python keyword or existing name |
| `__value` | Name mangling inside classes |
| `__value__` | Special system-defined name pattern; avoid creating your own |

In normal variable names, use a single leading underscore for "internal" names, a single trailing underscore to avoid naming conflicts, and avoid creating new names that both begin and end with double underscores.

````python
# Example: 3) _ And __ In Function Names
def _normalize_name(name):
    return name.strip().title()

def create_greeting(name):
    normalized_name = _normalize_name(name)
    return f"Hello, {normalized_name}!"

print(create_greeting("  alice  "))

first, _, third = ["A", "unused", "C"]
print("first:", first)
print("third:", third)

class_ = "Python fundamentals"
print("class name:", class_)

def Student():
    """Docstring!"""
    _internal_score = 95
    print("docstring:", Student.__doc__)
    print("dunder variable:", Student.__name__)
    print("internal score:", _internal_score)

# print("internal score:", _internal_score)  # This will raise a NameError because _internal_score is a local variable inside the Student function and is not accessible outside of it.
Student()
# Press shift+enter to run the code
````

# 5. Async And Exceptions

## 1) `async` And `await`

`async def` defines an asynchronous function.

`await` pauses until an asynchronous operation finishes.

Async functions are useful for waiting on I/O tasks such as network requests, file operations, or database calls.

⚠️ <strong>Notebook note:</strong> In Jupyter, top-level `await` is supported. In a regular Python script, you usually run async code with `asyncio.run(main())`.

````python
# Example: 1) async And await
import asyncio

async def fetch_score(name):
    await asyncio.sleep(0.1)
    return {"name": name, "score": 90}

result = await fetch_score("Alice")
print(result)

# Press shift+enter to run the code
````

## 2) Exception Handling With `try` And `except`

Exceptions are errors that happen while a program is running.

Use `try` and `except` to handle expected errors gracefully.

Common structure:

```python
try:
    risky_code
except SomeError:
    recovery_code
else:
    code_when_no_error
finally:
    code_that_always_runs
```

````python
# Example: 2) Exception Handling With try And except
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"
    else:
        return result
    finally:
        print("divide() finished")

print(divide(10, 2))
print(divide(10, 0))

# Press shift+enter to run the code
````

## 3) Custom Exceptions

A <strong>custom exception</strong> is your own error type.

Create one by defining a class that inherits from `Exception`.

Use custom exceptions when your program needs a clear domain-specific error.

````python
# Example: 3) Custom Exceptions
class InvalidScoreError(Exception):
    """Raised when a score is outside the valid range."""


def validate_score(score):
    if score < 0 or score > 100:
        raise InvalidScoreError("Score must be between 0 and 100.")
    return score

try:
    print(validate_score(95))
    print(validate_score(120))
except InvalidScoreError as error:
    print("Invalid score:", error)

# Press shift+enter to run the code
````

Remember these core ideas:

- Use functions to name and reuse behavior.
- Parameters belong to function definitions; arguments belong to function calls.
- Prefer returning values when you want reusable results.
- Be careful when changing mutable arguments inside a function.
- Use `*args`, `**kwargs`, and keyword-only arguments when function inputs need flexibility and clarity.
- Functions can be passed, returned, nested, decorated, and used to generate values lazily.
- Handle expected errors with `try` / `except`, and create custom exceptions when a domain-specific error improves clarity.

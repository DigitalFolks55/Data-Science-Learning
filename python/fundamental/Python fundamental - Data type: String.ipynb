# 3. Indexing & Slicing

## 1) Indexing

A string is an ordered sequence of characters.

Each character has an index which start from the begining (positive index) or from the end (negative index).

For `"Python"`:

| Character | P | y | t | h | o | n |
|---|---:|---:|---:|---:|---:|---:|
| Positive index | 0 | 1 | 2 | 3 | 4 | 5 |
| Negative index | -6 | -5 | -4 | -3 | -2 | -1 |

Python indexing starts from `0`.

````python
# Example: 1) Indexing
word = "Python"

print(word[0])   # first character
print(word[1])   # second character
print(word[-1])  # last character
print(word[-2])  # second from the end

# Press shift+enter to run the code
````

## 2) Slicing

Slicing extracts part of a string.
There are several ways to slice texts as below

```python
text[:]  # Extract all sequences from the start and the end.
text[start:]  # Extract sequences from the offset of start to the end.
test[:end]  # Extract sequence from the start to the offset of end (-1).
text[start:stop]  # Extract sequences from the offset of start to the offset of end (end-1).
text[start:stop:step]  # Extract sequences from the offset of start to the offset of end (end-1) with the interval of step. 
```

The `step` controls how many positions to move.

````python
# Example: 2) Slicing
word = "Python"

print(word[:])  # Python
print(word[3:])   # from index 3 to end
print(word[:3])   # from beginning to index 2
print(word[0:3])  # Pyt
print(word[1:4])  # yth
print(word[2:5:2])  # to

# Press shift+enter to run the code
````

# 4. String Immutability

Strings are <strong>immutable</strong>.

This means you cannot change a string object in place.

You can create a new string, but you cannot directly replace one character inside the existing string.

````python
# Example: 4. String Immutability
word = "Python"

# This will cause TypeError if uncommented.
# word[0] = "J"

new_word = "J" + word[1:]

print("Original:", word)
print("New:", new_word)

# Press shift+enter to run the code
````

## Why This Matters

Many string methods return a new string.

They do not modify the original string.

````python
# Example: Why This Matters
text = "python"
upper_text = text.upper()

print("Original:", text)
print("New:", upper_text)

# Press shift+enter to run the code
````

# 6. String Methods

String methods are functions attached to string objects.

Most string methods do **not** change the original string. They usually return a new string, because strings are immutable.

Use string methods when you need to:

- change letter case
- remove unnecessary spaces or symbols
- search for text
- replace text
- split text into smaller parts
- combine text values
- validate text content
- align text for display

Common methods:

| Method | Meaning |
|---|---|
| `.upper()` | Convert all letters to uppercase |
| `.lower()` | Convert all letters to lowercase |
| `.strip()` | Remove whitespace, or specified characters, from both ends |
| `.lstrip()` | Remove whitespace, or specified characters, from the left end |
| `.rstrip()` | Remove whitespace, or specified characters, from the right end |
| `.replace(old, new)` | Replace matching text with new text |
| `.startswith(text)` | Check whether a string starts with specific text |
| `.endswith(text)` | Check whether a string ends with specific text |
| `.find(text)` | Find the first position of text; returns `-1` if not found |
| `.rfind(text)` | Find the last position of text; returns `-1` if not found |
| `.index(text)` | Find the first position of text; raises `ValueError` if not found |
| `.rindex(text)` | Find the last position of text; raises `ValueError` if not found |
| `.count(text)` | Count how many times text appears |
| `.split()` | Split a string into a list of strings |
| `.join()` | Combine multiple strings into one string |
| `.capitalize()` | Capitalize the first character and lowercase the rest |
| `.title()` | Capitalize the first character of each word |
| `.swapcase()` | Swap uppercase letters to lowercase and lowercase letters to uppercase |
| `.center(width)` | Center text inside a field of a given width |
| `.ljust(width)` | Left-align text inside a field of a given width |
| `.rjust(width)` | Right-align text inside a field of a given width |

⚠️ <strong>Important:</strong> `.find()` and `.index()` look similar, but they behave differently when text is missing. `.find()` returns `-1`, while `.index()` raises an error.

````python
# Basic case conversion and text replacement
text = "python programming"

# Case methods return new strings.
print("upper:", text.upper())
print("lower:", "PYTHON".lower())
print("capitalize:", text.capitalize())
print("title:", text.title())
print("swapcase:", "PyThOn".swapcase())

# replace() returns a new string with matching text replaced.
message = "I like Java"
print("replace:", message.replace("Java", "Python"))

# count() counts matching text.
fruit = "banana"
print("count:", fruit.count("a"))

# Press shift+enter to run the code
````

````python
# Example: 6. String Methods
print(text.strip().startswith("Python"))
print(text.strip().endswith("ing"))

# Press shift+enter to run the code
````

## Method Chaining

Because many string methods return a new string, you can chain methods.

Method chaining means calling one method after another:

```python
raw_text.strip().lower().replace("python", "data")
```

Python evaluates the chain from left to right:

1. `strip()` removes outside spaces.
2. `lower()` converts the cleaned text to lowercase.
3. `replace()` replaces matching text.

⚠️ <strong>Tip:</strong> Use chaining when each step is simple. If the chain becomes hard to read, split it into multiple lines or variables.

````python
# Method chaining keeps simple cleaning steps compact.
raw_text = "  PYTHON programming  "
clean_text = raw_text.strip().lower().replace("python", "data")

print("Original:", repr(raw_text))
print("Cleaned:", clean_text)

# Press shift+enter to run the code
````

## Check String Content

Some string methods return Boolean values. They are useful for validation before converting, saving, or processing text.

| Method | Example meaning |
|---|---|
| `.isalpha()` | `True` if all characters are alphabetic and the string is not empty |
| `.isdigit()` | `True` if all characters are digit characters and the string is not empty |
| `.isalnum()` | `True` if all characters are alphabetic or digit characters and the string is not empty |
| `.isspace()` | `True` if all characters are whitespace and the string is not empty |
| `.islower()` | `True` if all cased letters are lowercase |
| `.isupper()` | `True` if all cased letters are uppercase |

⚠️ <strong>Watch out:</strong> These methods can be strict. For example, `"Python 3".isalnum()` is `False` because the space is not alphabetic or numeric.

````python
# Content-check methods return True or False.
samples = ["Python", "123", "Python3", "Python 3", "   ", "PYTHON", "python", ""]

for sample in samples:
    print("Value:", repr(sample))
    print("isalpha:", sample.isalpha())
    print("isdigit:", sample.isdigit())
    print("isalnum:", sample.isalnum())
    print("isspace:", sample.isspace())
    print("islower:", sample.islower())
    print("isupper:", sample.isupper())
    print("---")

# Press shift+enter to run the code
````

## Practical Cleaning Example

A common data-cleaning task is to remove extra spaces and standardize text.

In the next cell, the raw text contains names separated by commas. Some names have extra spaces and inconsistent capitalization.

The cleaning process is:

1. Use `.split(",")` to separate the names.
2. Use `.strip()` to remove extra spaces around each name.
3. Use `.title()` to standardize capitalization.
4. Use `.join()` to combine the cleaned names into one readable string.

````python
# Clean comma-separated names.
raw_names = "  alice, BOB,   charlie,David  "

names = raw_names.split(",")
clean_names = [name.strip().title() for name in names]
result = ", ".join(clean_names)

print("Raw split result:", names)
print("Clean list:", clean_names)
print("Final text:", result)

# Press shift+enter to run the code
````

# 7. Formatting Strings

String formatting creates readable messages from values.

Python has several formatting styles:

| Style | Example | Recommendation |
|---|---|---|
| f-string | `f"Hello, {name}"` | Best default for modern Python |
| `.format()` | `"Hello, {}".format(name)` | Common in older Python code and reusable templates |
| `%` formatting | `"Hello, %s" % name` | Old style; useful to recognize in legacy code |

The most common modern style is an <strong>f-string</strong>, but you will still see `%` formatting and `.format()` as usual.

## 1) f-string formatting

### Numeric formatting with f-string

````python
# Example: Numeric formatting with f-string
price = 1234.5678
rate = 0.03456

print(f"Price: {price:.2f}")       # 2 decimal places
print(f"Price: {price:,.2f}")      # comma separator + 2 decimals
print(f"Rate: {rate:.2%}")         # percentage
print(f"Right aligned: {price:>12.2f}")

# Press shift+enter to run the code
````

### Debug-style f-string

`{variable=}` is useful when checking code.

````python
# Example: Debug-style f-string
language = "Python"
version = 3.12

print(f"{language=}")
print(f"{version=}")

# Press shift+enter to run the code
````

### Tips on `f-string` Formatting

1. A colon `:` is used inside `{}` to control formatting.
   - `{value:10}` means the value uses a minimum width of 10 characters.
   - `{value:<10}` means left alignment.
   - `{value:>10}` means right alignment.
   - `{value:^10}` means center alignment.

2. A number such as `10.3` can be used to control width and precision.
   - For strings, the first number is the minimum width, and the second number is the maximum number of characters.
   - For floating-point numbers, the first number is the minimum width, and the second number is the number of digits after the decimal point.

3. Extra symbols can be used for number formatting.
   - `{value:+.2f}` shows the sign for both positive and negative numbers.
   - `{value:,}` separates thousands with commas.
   - `{value:.2%}` displays a number as a percentage with 2 digits after the decimal point.

````python
# Insert expressions
price = 100
quantity = 3
result = f"Total price is {price * quantity}"
print(result)

# Minimum width
text = "Python"
result = f"{text:10}"
print(result)

# Left alignment
text = "Python"
result = f"{text:<10}"
print(result)

# Right alignment
text = "Python"
result = f"{text:>10}"
print(result)

# Center alignment
text = "Python"
result = f"{text:^10}"
print(result)

# String width and precision
text = "Python"
result = f"{text:10.3}"
print(result)

# Floating-point precision
value = 3.14159
result = f"{value:.2f}"
print(result)

# Floating-point width and precision
value = 3.14159
result = f"{value:10.2f}"
print(result)

# Show sign
value = 3.14159
result = f"{value:+.2f}"
print(result)

# Thousands separator
value = 1000000
result = f"{value:,}"
print(result)

# Percentage formatting
rate = 0.12345
result = f"{rate:.2%}"
print(result)

# Press shift+enter to run the code
````

## 2) `%` formatting

### `%` Formatting

| Placeholder | Meaning | Example value |
|---|---|---|
| `%s` | String or string representation | `"Alice"` |
| `%d` | Integer (Decimal) | `25` |
| `%x` | Hexa | `%2a` |
| `%o` | Octa | `%52` |
| `%f` | Floating-point number | `87.500000` |
| `%.2f` | Floating-point number with 2 decimals | `87.50` |
| `%e` | Exponential floating-poing number | `7.02e+00` |
| `%%` | Literal value of % | `%` |

````python
# % formatting
name = "Alice"
age = 25
score = 87.5

print("Name: %s" % name)
print("Age: %d" % age)
print("Score: %.2f" % score)
print("%s is %d years old. Score: %.1f" % (name, age, score))

# Press shift+enter to run the code
````

### Tips of `%` Formatting

1. A formatting placeholder begins with `%`.

2. `+` or `-` can follow `%`.

   - `+` shows the sign for positive numbers.
   - `-` means left alignment.

3. A number such as `10.3` can be used to control width and precision.

   - For strings, the first number is the minimum width, and the second number is the maximum number of characters.
   - For floating-point numbers, the first number is the minimum width, and the second number is the number of digits after the decimal point.

````python
# Example of string formatting with % operator
string = "Himeji Castle"

print("%s" % string)
print("%20s" % string)  # right-aligned in 20 spaces
print("%-20s" % string) # left-aligned in 20 spaces
print("%.5s" % string)  # first 5 characters
print("%20.5s" % string) # first 5 characters, right-aligned in 20 spaces

# Press shift+enter to run the code
````

````python
# Example of float formatting with % operator
float_value = 1234.56789

print("%f" % float_value)
print("%10.2f" % float_value)  # right-aligned in 10 spaces, 2 decimal places
print("%-10.2f" % float_value) # left-aligned in 10 spaces, 2 decimal places
print("%.2f" % float_value)    # 2 decimal places
print("%10.2f" % float_value)  # right-aligned in 10 spaces, 2 decimal places

# Press shift+enter to run the code
````

## 3) `.format()` formatting


`.format()` uses `{}` placeholders inside the string.

You can use:

- empty placeholders: `{}`
- numbered placeholders: `{0}`, `{1}`
- named placeholders: `{name}`, `{score}`
- formatting rules after `:` such as `{score:.2f}`

````python
# str.format()
name = "Alice"
age = 25
score = 87.5

print("Name: {}".format(name))
print("{} is {} years old. Score: {}".format(name, age, score))
print("{0} scored {2}. {0} is {1} years old.".format(name, age, score))
print("{name} is {age} years old. Score: {score:.2f}".format(name=name, age=age, score=score))

# Press shift+enter to run the code
````

### Tips on `{}` and `.format()` Formatting

1. A formatting placeholder is written with `{}`.

2. Values are passed to the placeholder by using `.format()`.
   - `{}` inserts values in order.
   - `{0}`, `{1}`, `{2}` insert values by position.
   - `{name}` inserts values by keyword name.

3. A colon `:` is used inside `{}` to control formatting.
   - `{:10}` means the value uses a minimum width of 10 characters.
   - `{:<10}` means left alignment.
   - `{:>10}` means right alignment.
   - `{:^10}` means center alignment.

4. A number such as `10.3` can be used to control width and precision.
   - For strings, the first number is the minimum width, and the second number is the maximum number of characters.
   - For floating-point numbers, the first number is the minimum width, and the second number is the number of digits after the decimal point.

5. Extra symbols can be used for number formatting.

   - `{:+.2f}` shows the sign for both positive and negative numbers.
   - `{:,}` separates thousands with commas.
   - `{:.2%}` displays a number as a percentage with 2 digits after the decimal point.

````python
# Minimum width
result = "{:10}".format("Python")
print(result)
# Left alignment
result = "{:<10}".format("Python")
print(result)
# Right alignment
result = "{:>10}".format("Python")
print(result)
# Center alignment
result = "{:^10}".format("Python")
print(result)
# String width and precision
result = "{:10.3}".format("Python")
print(result)
# Floating-point width and precision
result = "{:10.2f}".format(3.14159)
print(result)
# Show sign
result = "{:+.2f}".format(3.14159)
print(result)
# Thousands separator
result = "{:,}".format(1000000)
print(result)
# Percentage formatting
result = "{:.2%}".format(0.12345)
print(result)

# Press shift+enter to run the code
````

# 10. Key Terms

| Term | Meaning |
|---|---|
| String | Text data stored as a sequence of characters |
| `str` | Python's string type |
| Empty string | A string with no characters: `""` |
| Index | A character position; starts at `0` |
| Negative index | Counts from the end; `-1` is the last character |
| Slicing | Extracting part of a string with `text[start:stop:step]` |
| Immutability | Strings cannot be changed in place |
| Concatenation | Combining strings with `+` |
| Membership | Checking text with `in` |
| String method | A function attached to a string, such as `.strip()` |
| Method chaining | Calling methods in sequence, such as `text.strip().lower()` |
| `split()` | Converts a string into a list of strings |
| `join()` | Combines strings into one string |
| f-string | Modern formatting style: `f"Hello, {name}"` |
| `%` formatting | Older formatting style using `%s`, `%d`, and `%f` |
| `.format()` | Formatting style using `{}` placeholders and `.format()` |
| Escape character | A backslash sequence such as `\n`, `\t`, or `\\` |
| Raw string | A string prefixed with `r` that keeps backslashes as written |

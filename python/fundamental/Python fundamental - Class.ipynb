# 1. Class Basics

## 1) What Is An Instance Object?

An <strong>object</strong> is a value that has data and behavior.

An <strong>instance</strong> is an object created from a class.

| Word | Meaning | Example |
|---|---|---|
| Class | Blueprint for objects | `Student` |
| Instance / object | Concrete value created from a class | `alice = Student()` |
| Attribute | Data stored on an object | `alice.name` |
| Method | Function attached to an object or class | `alice.introduce()` |

A class is like a blueprint. An instance is the actual object built from that blueprint.

````python
# Example: 1) What Is An Instance Object?
text = "Python"
numbers = [1, 2, 3]

print(type(text))
print(type(numbers))
print(text.upper())
print(numbers.append)

# Press shift+enter to run the code
````

## 2) Class Definition

Use `class` to define a customized class.

Basic syntax:

```python
class ClassName:
    code_inside_class
```

Class names usually use <strong>PascalCase</strong>, such as `StudentProfile` or `BankAccount`.

````python
# Example: 2) Class Definition
class Student:
    pass

alice = Student()
ben = Student()

print(alice)
print(ben)
print(type(alice))
print(alice is ben)

# Press shift+enter to run the code
````

## 3) Attributes And Methods

An <strong>attribute</strong> stores data.

A <strong>method</strong> stores behavior.

Attributes and methods are accessed with dot notation.

````python
# Example: 3) Attributes And Methods
class Student:
    def introduce(self):
        return f"Hello, I am {self.name}."

alice = Student()

# If no attributes are defined, then it will raise up errors.
# print(alice.name)  # This will raise up Attribute error!
# print(alice.score)  # This will raise up Attribute error!
# print(alice.introduce())   # This will raise up Attribute error due to alice.name (self.name) is not defined yet.

alice.name = "Alice"
alice.score = 90

print(alice.name)
print(alice.score)
print(alice.introduce())

# Press shift+enter to run the code
````

## 4) Initialization With `__init__()`

`__init__()` initializes a new instance after it is created.  
This is not a mandatory method for any class.

Use it to set the starting attributes for each object.

````python
# Example: 4) Initialization With __init__()
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
        print(f"attributes are name: {self.name} and score: {self.score}")

    def introduce(self):
        return f"{self.name} scored {self.score}."

alice = Student("Alice", 90)

print(alice.introduce())

# Press shift+enter to run the code
````

What happened above?
1) Defined "Student" class.
2) Load "Student" class and new instance was created.
3) 2 variables; name and score were passed to the class and `__init__()` method was loaded.
4) Store 1st variable to attribute "name" and 2nd variable to attribute "score".
5) Return new instance which has 2 attributes.
6) Assign new instance to alice.
7) Execute "introduce" method.

## 5) `self`

`self` refers to the current instance.

When you call `alice.introduce()`, Python automatically passes `alice` as the first argument to the method.

`self` is a convention, but you should use it because Python developers expect it.

````python
# Example: 5) self
class Counter:
    def __init__(self):
        self.count = 0

    def increase(self):
        self.count += 1
        return self.count

counter = Counter()
print(counter.increase())
print(counter.increase())
print(counter.count)

# Press shift+enter to run the code
````

# 2. Access To Attributes

## 1) Direct Attribute Access

Direct attribute access uses dot notation.

```python
object.attribute
object.attribute = value
```

This is simple and common in Python.

````python
# Example: 1) Direct Attribute Access
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

student = Student("Alice", 90)
print(student.name)

student.score = 95
print(student.score)

# Press shift+enter to run the code
````

## 2) Getter And Setter Methods

A <strong>getter</strong> reads an attribute.

A <strong>setter</strong> updates an attribute, often with validation.

Getter and setter methods are useful when direct access needs rules.

````python
# Example: 2) Getter And Setter Methods
class Student:
    def __init__(self, name, score):
        self.name = name
        self.set_score(score)

    def get_score(self):
        return self._score

    def set_score(self, score):
        if score < 0 or score > 100:
            raise ValueError("Score must be between 0 and 100.")
        self._score = score

student = Student("Alice", 90)
print(student.get_score())
student.set_score(95)
print(student.get_score())

# Press shift+enter to run the code
````

## 3) `property`

`property` lets you use attribute-style access while still running getter and setter logic.

This gives a clean interface and keeps validation inside the class.

````python
# Example: 3) property
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    @property
    def score(self):
        return self._score

    @score.setter
    def score(self, value):
        if value < 0 or value > 100:
            raise ValueError("Score must be between 0 and 100.")
        self._score = value

student = Student("Alice", 90)
print(student.score)
student.score = 98
print(student.score)
# student.score = 200  # This will raise value error which defined in a setter method.

# Press shift+enter to run the code
````

````python
class CircularArea:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return 3.14*self.radius**2

area = CircularArea(10)
print(area.radius)
print(area.area)

# Access to attribute and overwrite
area.radius = 5
print(area.radius)
print(area.area)
# area.area = 100  # This will raise Attribute error and direct access is not available.

# Press shift+enter to run the code
````

## 4) Name Mangling

Names that start with double underscores, such as `__balance`, are name-mangled by Python.

Name mangling helps avoid accidental name conflicts in subclasses. It is not strong security.

````python
# Example: 4) Name Mangling
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        return self.__balance

account = BankAccount(1000)
print(account.show_balance())
print(account.__dict__)
# print(account.__balance)  # This would raise AttributeError

# Press shift+enter to run the code
````

## 5) Class Attribute And Instance Attribute

A <strong>class attribute</strong> belongs to the class and is shared by instances.

An <strong>instance attribute</strong> belongs to one object.

````python
# Example: 5) Class Attribute And Instance Attribute
class Student:
    school = "Data School"  # class attribute

    def __init__(self, name):
        self.name = name  # instance attribute

alice = Student("Alice")
ben = Student("Ben")

print(alice.school, alice.name)
print(ben.school, ben.name)

Student.school = "Python School"
print(alice.school)
print(ben.school)

# Press shift+enter to run the code
````

# 3. Method Types

## 1) Instance Method

An <strong>instance method</strong> receives the instance as `self`.

Use instance methods when the behavior needs instance data.

````python
# Example: 1) Instance Method
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def passed(self):
        return self.score >= 60

student = Student("Alice", 90)
print(student.passed())

# Press shift+enter to run the code
````

## 2) Class Method

A <strong>class method</strong> receives the class as `cls`.

Use `@classmethod` when the method works with the class itself or creates instances in an alternate way.

````python
# Example: 2) Class Method
class Student:
    school = "Data School"

    def __init__(self, name, score):
        self.name = name
        self.score = score

    @classmethod
    def from_text(cls, text):
        name, score = text.split(",")
        return cls(name, int(score))

student = Student.from_text("Alice,90")
print(student.name, student.score)

# Press shift+enter to run the code
````

## 3) Static Method

A <strong>static method</strong> does not receive `self` or `cls` automatically.

Use `@staticmethod` when the behavior belongs near the class but does not need instance or class data.

````python
# Example: 3) Static Method
class ScoreTool:
    @staticmethod
    def is_valid(score):
        return 0 <= score <= 100

print(ScoreTool.is_valid(95))
print(ScoreTool.is_valid(120))

# Press shift+enter to run the code
````

## 4) Special Method And Magic Method

Special methods have double underscores before and after the name.

They are also called <strong>magic methods</strong> or <strong>dunder methods</strong>.

|Method|Description|
|---|---|
|`__eq__(self, other)`|self == other|
|`__ne__(self, other)`|self != other|
|`__lt__(self, other)`|self < other|
|`__gt__(self, other)`|self > other|
|`__le__(self, other)`|self <= other|
|`__ge__(self, other)`|self >= other|
|`__add__(self, other)`|self + other|
|`__sub__(self, other)`|self - other|
|`__mul__(self, other)`|self * other|
|`__floordiv__(self, other)`|self // other|
|`__truediv__(self, other)`|self / other|
|`__mod__(self, other)`|self % other|
|`__pow__(self, other)`|self ** other|
|`__str__(self)`|str(self)|
|`__repr__(self)`|repr(self)|
|`__len__(self)`|len(self)|

````python
# Example: 4) Special Method And Magic Method
class Course:
    def __init__(self, title, students):
        self.title = title
        self.students = students

    def __str__(self):
        return self.title

    def __len__(self):
        return len(self.students)

course = Course("Python", ["Alice", "Ben", "Chloe"])
print(str(course))
print(len(course))

# Press shift+enter to run the code
````

# 4. Inheritance

## 1) Inherit From Parent Class

Inheritance lets a child class reuse behavior from a parent class.

```python
class Child(Parent):
    ...
```

Use inheritance when one class is a specialized version of another class.

````python
# Example: 1) Inherit From Parent Class
class Employee:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I am {self.name}."

class DataAnalyst(Employee):
    pass

analyst = DataAnalyst("Alice")
print("DataAnlyst is a subclass of Employee?", issubclass(DataAnalyst, Employee))
print(analyst.introduce())

# Press shift+enter to run the code
````

## 2) Override Method

A child class can override a parent method by defining a method with the same name.

````python
# Example: 2) Override Method
class Employee:
    def work(self):
        return "Working"

class DataAnalyst(Employee):
    def work(self):
        return "Analyzing data"

employee = Employee()
analyst = DataAnalyst()

print(employee.work())
print(analyst.work())

# Press shift+enter to run the code
````

## 3) Add Method

A child class can add new methods that the parent class does not have.

````python
# Example: 3) Add Method
class Employee:
    def __init__(self, name):
        self.name = name

class DataAnalyst(Employee):
    def analyze(self, dataset):
        return f"{self.name} analyzes {dataset}."

analyst = DataAnalyst("Alice")
print(analyst.analyze("sales data"))

# Press shift+enter to run the code
````

## 4) `super()`

`super()` calls a method from the parent class.

It is commonly used inside `__init__()` so the child can reuse parent initialization and add more attributes.

````python
# Example: 4) super()
class Employee:
    def __init__(self, name):
        self.name = name

class DataAnalyst(Employee):
    def __init__(self, name, tool):
        super().__init__(name)  # Load __init__ method from a parent class and pass "name" attribute
        self.tool = tool

    def introduce(self):
        return f"I am {self.name}, and I use {self.tool}."

analyst = DataAnalyst("Alice", "Python")
print(analyst.introduce())

# Press shift+enter to run the code
````

## 5) Multiple Inheritance

A class can inherit from more than one parent class.

Python searches methods using the method resolution order, also called <strong>MRO</strong>.

Use multiple inheritance carefully because it can make relationships harder to understand.

````python
# Example: 5) Multiple Inheritance
class Logger:
    def log(self, message):
        print("LOG:", message)

class Validator:
    def is_valid_score(self, score):
        return 0 <= score <= 100

class ScoreService(Logger, Validator):
    def save_score(self, score):
        if self.is_valid_score(score):
            self.log(f"Saved score {score}")
        else:
            self.log("Invalid score")

service = ScoreService()
service.save_score(95)
print(ScoreService.__mro__)

# Press shift+enter to run the code
````

If you have same method to be overwritten in child classes and new class inherit them together, then a method in first place (you can check it with mro) will be applied.

````python
class Base:
    def announce(self):
        return "Base"

class First(Base):
    def announce(self):
        return "First"

class Second(Base):
    def announce(self):
        return "Second"

class FirstSecond(First, Second):
    pass

class SecondFirst(Second, First):
    pass

# Check mro
print(FirstSecond.mro())
print(SecondFirst.mro())

# Check which method will be applied
print(FirstSecond().announce())  # announce in First is a first in place
print(SecondFirst().announce())  # announce in Second is a first in place

# Press shift+enter to run the code
````

## 6) Mixin

A <strong>mixin</strong> is a small class that adds one focused behavior.

Mixins are usually not used alone. They are combined with another class.

````python
# Example: 6) Mixin
class JsonMixin:
    def to_dict(self):
        return self.__dict__

class Student(JsonMixin):
    def __init__(self, name, score):
        self.name = name
        self.score = score

student = Student("Alice", 90)
print(student.to_dict())

# Press shift+enter to run the code
````

# 5. Object-Oriented Patterns

## 1) Duck Typing

Duck typing means Python cares about what an object can do, not only what class it belongs to.

If different objects have the same expected method, they can be used in the same code.

````python
# Example: 1) Duck Typing
class CsvExporter:
    def export(self):
        return "Exporting CSV"

class JsonExporter:
    def export(self):
        return "Exporting JSON"

def run_export(exporter):
    print(exporter.export())

run_export(CsvExporter())
run_export(JsonExporter())

# Press shift+enter to run the code
````

## 2) Aggregation

Aggregation means one object uses another object, but the used object can exist independently.

Example: A course has a teacher, but the teacher can exist without that course.

````python
# Example: 2) Aggregation
class Teacher:
    def __init__(self, name):
        self.name = name

class Course:
    def __init__(self, title, teacher):
        self.title = title
        self.teacher = teacher

teacher = Teacher("Dr. Kim")
course = Course("Python", teacher)

print(course.title)
print(course.teacher.name)

# Press shift+enter to run the code
````

## 3) Composition

Composition means one object owns another object as part of itself.

Example: A report creates and owns its pages.

````python
# Example: 3) Composition
class Page:
    def __init__(self, number):
        self.number = number

class Report:
    def __init__(self, title, page_count):
        self.title = title
        self.pages = [Page(number) for number in range(1, page_count + 1)]

report = Report("Monthly Sales", 3)
print(report.title)
print([page.number for page in report.pages])

# Press shift+enter to run the code
````

## 4) Instance Object Vs Others: How To Decide

Classes are powerful, but they are not always necessary.

| Choice | Use when |
|---|---|
| Function | You need one clear action or transformation |
| Dictionary | You need simple labeled data |
| Named tuple | You need a lightweight immutable record |
| Data class | You need a simple class mainly for storing data |
| Regular class | You need data plus behavior, validation, state changes, or extension |

Start simple. Use a class when the data and behavior naturally belong together.

````python
# Example: 4) Instance Object Vs Others: How To Decide
def calculate_tax(price, tax_rate):
    return price * tax_rate

class Invoice:
    def __init__(self, price, tax_rate):
        self.price = price
        self.tax_rate = tax_rate

    def total(self):
        return self.price + calculate_tax(self.price, self.tax_rate)

print(calculate_tax(1000, 0.1))
invoice = Invoice(1000, 0.1)
print(invoice.total())

# Press shift+enter to run the code
````

# 6. Structured Records

## 1) Named Tuple

A <strong>named tuple</strong> is a lightweight immutable record with named fields.

Use it when you want tuple-like data that is easier to read.

````python
# Example: 1) Named Tuple
from collections import namedtuple

StudentRecord = namedtuple("StudentRecord", ["name", "score"])
student = StudentRecord("Alice", 90)

print(student)
print(student.name)
print(student.score)

# Press shift+enter to run the code
````

## 2) Data Class

A <strong>data class</strong> creates common class methods automatically.

Use `@dataclass` for classes that mainly store data.

````python
# Example: 2) Data Class
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    score: int

    def passed(self):
        return self.score >= 60

student = Student("Alice", 90)
print(student)
print(student.passed())

# Press shift+enter to run the code
````

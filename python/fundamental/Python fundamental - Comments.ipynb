# 1. Comments With `#`

## 1) What Is A Comment?

A <strong>comment</strong> is text written for humans, not for Python.

Python ignores comments when running code.

Use comments to explain:

- why code exists
- assumptions or constraints
- non-obvious logic
- important warnings

Do not use comments to repeat obvious code.

````python
# Example: full-line comments
# This variable stores a student's score.
score = 85

# This condition checks whether the score is passing.
if score >= 60:
    print("Passed")
else:
    print("Failed")

# Press shift+enter to run the code
````

## 2) Full-Line Comments

A full-line comment starts with `#`.

```python
# This is a full-line comment.
```

Python skips the whole line.

````python
# Example: use full-line comments to separate code steps
# Step 1: Prepare input values
price = 1200
tax_rate = 0.10

# Step 2: Calculate total price
total_price = price * (1 + tax_rate)

# Step 3: Display the result
print(total_price)

# Press shift+enter to run the code
````

## 3) Inline Comments

An inline comment appears after code on the same line.

```python
score = 85  # This is an inline comment.
```

Use inline comments sparingly. They are helpful for short explanations, units, warnings, or assumptions.

````python
# Example: inline comments
speed = 80  # km per hour
time = 2    # hours

distance = speed * time
print("Distance:", distance, "km")

# Press shift+enter to run the code
````

## 4) Good And Bad Comments

Good comments explain <strong>why</strong>, not just <strong>what</strong>.

| Comment | Quality | Reason |
|---|---|---|
| `# Add 1 to count` | Weak | The code already says this |
| `# Start from 1 because invoice numbers are customer-facing` | Better | Explains the reason |
| `# Temporary workaround for missing email field` | Useful | Explains an unusual decision |

````python
# Example: explain why, not only what
invoice_number = 1  # Start from 1 because invoice numbers are customer-facing.

invoice_number += 1
print("Next invoice:", invoice_number)

# Press shift+enter to run the code
````

# 2. Backslash `\`

## 1) What Does Backslash Do?

A backslash `\` at the end of a line can continue one Python statement onto the next line.

This is called <strong>explicit line continuation</strong>.

```python
total = 10 + \
        20
```

⚠️ <strong>Important:</strong> The backslash must be the final character on the line. Do not put spaces or comments after it.

````python
# Example: explicit line continuation with backslash
total = 10 + \
        20 + \
        30

print(total)

# Press shift+enter to run the code
````

## 2) Prefer Parentheses For Long Expressions

Backslash works, but parentheses are usually cleaner and safer.

Python automatically continues lines inside `()`, `[]`, and `{}`.

````python
# Example: implicit line continuation with parentheses
total = (
    10
    + 20
    + 30
)

print(total)

names = [
    "Alice",
    "Ben",
    "Chloe",
]

print(names)

# Press shift+enter to run the code
````

## 3) Backslash Is Not A Comment

A backslash does not create a comment. It changes how Python reads the line.

Use `#` for comments. Use `\` only when you need explicit line continuation.

````python
# Example: backslash continues a statement, not a comment
message = "Hello, " \
          "Python"

print(message)

# Press shift+enter to run the code
````

# 3. Triple Quotes `"""`

## 1) Triple-Quoted Strings

Triple quotes create a multi-line string.

You can use triple double quotes or triple single quotes.

Triple-quoted strings are useful for:

- long text
- multi-line messages
- documentation strings, called docstrings

````python
# Example: multi-line string with triple quotes
message = """
Python is readable.
Python is flexible.
Python is widely used.
"""

print(message)

# Press shift+enter to run the code
````

## 2) Triple Quotes Are Not Exactly Comments

Triple-quoted text is a string object.

Python does not ignore it in the same way it ignores `#` comments.

If a triple-quoted string is not assigned to a variable and is not a docstring, it may behave like a harmless standalone string, but it is still a string expression.

For real comments, use `#`.

````python
# Example: standalone triple-quoted string
"""
This looks like a multi-line comment.
Python treats it as a string expression.
Use # for normal comments.
"""

print("Code still runs after the standalone string.")

# Press shift+enter to run the code
````

## 3) Docstrings

A <strong>docstring</strong> is a triple-quoted string placed at the beginning of a function, class, or module.

Docstrings explain what the function, class, or module does.

You can read a function docstring with `help()` or `function.__doc__`.

````python
# Example: function docstring
def calculate_total(price, tax_rate):
    """
    Calculate the total price after tax.

    Parameters:
        price: Original price before tax.
        tax_rate: Tax rate as a decimal number.
    """
    return price * (1 + tax_rate)

print(calculate_total(1200, 0.10))
print(calculate_total.__doc__)

# Press shift+enter to run the code
````

## 4) Single-Line Docstrings

Short functions can use one-line docstrings.

````python
# Example: one-line docstring
def greet(name):
    """Return a greeting message for a name."""
    return f"Hello, {name}!"

print(greet("Alice"))
print(greet.__doc__)

# Press shift+enter to run the code
````

# 4. Practical Comment Style

## 1) When Should You Add Comments?

Add comments when they reduce confusion.

Good places for comments:

- before a complex block
- beside a non-obvious value
- near a temporary workaround
- to explain business rules
- to warn about side effects

Avoid comments that simply repeat the code.

````python
# Example: explain a business rule
purchase_amount = 6000
is_member = True

# Members get free shipping when purchase amount is 5000 or more.
if is_member and purchase_amount >= 5000:
    shipping_fee = 0
else:
    shipping_fee = 700

print("Shipping fee:", shipping_fee)

# Press shift+enter to run the code
````

## 2) Commenting Out Code

Sometimes learners temporarily comment out code while testing.

This is useful during practice, but production code should not keep large blocks of unused commented-out code.

````python
# Example: temporarily comment out code
name = "Alice"

print("Hello", name)
# print("This line is temporarily disabled")
print("Done")

# Press shift+enter to run the code
````

## 3) Comment Checklist

Before writing a comment, ask:

1. Is the code already clear?
2. Does the comment explain why or warn about something important?
3. Will this comment still be true if the code changes?
4. Would a better variable name remove the need for this comment?

# 6. Key Terms

| Term | Meaning |
|---|---|
| Comment | Text for humans that Python ignores |
| `#` | Starts a comment in Python |
| Full-line comment | A whole line used as a comment |
| Inline comment | A comment after code on the same line |
| Backslash `\` | Explicitly continues one statement onto the next line |
| Implicit continuation | Line continuation inside `()`, `[]`, or `{}` |
| Triple-quoted string | A multi-line string written with triple quotes |
| Docstring | A triple-quoted string that documents a function, class, or module |
| `__doc__` | Attribute that stores an object's docstring |
| Commenting out | Temporarily disabling code by adding `#` |

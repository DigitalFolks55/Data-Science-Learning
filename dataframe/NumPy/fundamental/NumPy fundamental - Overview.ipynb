# 1. What Is NumPy?

NumPy is a Python library for numerical and scientific computing. The name comes from Numerical Python.

NumPy provides:

| Feature | Meaning |
|---|---|
| `ndarray` | N-dimensional array object for storing regular numerical data |
| Array operations | Fast mathematical, logical, statistical, and shape manipulation routines |
| Numerical tools | Linear algebra, random sampling, Fourier transforms, sorting, and more |
| Ecosystem foundation | Array structures used by many data science and scientific Python libraries |

In beginner terms, NumPy helps Python work with many numbers efficiently.

```python
# Import NumPy and check the installed version
import numpy as np

print("NumPy version:", np.__version__)
print("Main array type:", np.ndarray)

# Press shift+enter to run the code
```

### 🗓️ Where NumPy Fits

NumPy is often used between basic Python and higher-level data science libraries.

| Layer | Example tool | Typical role |
|---|---|---|
| General Python | `list`, `tuple`, `dict` | Flexible programming structures |
| Numerical arrays | NumPy | Fast numerical operations on shaped data |
| Labeled tables | pandas, Polars | Rows, columns, labels, joins, and grouped analysis |
| Modeling and science | scikit-learn, SciPy | Machine learning and scientific computing workflows |

Learning NumPy makes later libraries easier because many of them use array concepts such as shape, dtype, axis, and vectorized operations.

# 2. Create A First `ndarray`

The central NumPy object is the N-dimensional array, usually called an `ndarray`.

An array has metadata that describes its structure:

| Attribute | Meaning | Example |
|---|---|---|
| `.shape` | Length of each dimension | `(5,)` for a one-dimensional array with 5 values |
| `.ndim` | Number of dimensions | `1` for a vector, `2` for a matrix |
| `.size` | Total number of values | `5` |
| `.dtype` | Data type stored in the array | `int64`, `float64`, `bool`, and others |

```python
# Create a first NumPy array and inspect it
python_list = [10, 20, 30, 40, 50]
numpy_array = np.array(python_list)

print("Python list:", python_list)
print("NumPy array:", numpy_array)
print("Type:", type(numpy_array))
print("Shape:", numpy_array.shape)
print("Dimensions:", numpy_array.ndim)
print("Size:", numpy_array.size)
print("Data type:", numpy_array.dtype)

# Press shift+enter to run the code
```

### 🗓️ Python List Versus NumPy Array

Python lists are general-purpose containers. They can hold mixed data types and are designed for flexibility.

NumPy arrays are designed for numerical work. They usually store one data type in a compact structure, which makes mathematical operations more consistent and often faster.

| Container | Main purpose | Typical data | Strength |
|---|---|---|---|
| Python list | General collection | Mixed or flexible data | Easy and flexible |
| NumPy array | Numerical computing | Same-type numeric data | Fast array operations |

```python
# Compare list behavior and array behavior
python_list = [1, 2, 3, 4, 5]
numpy_array = np.array([1, 2, 3, 4, 5])

print("python_list * 2:", python_list * 2)
print("numpy_array * 2:", numpy_array * 2)
print()
print("List comprehension for element-wise multiplication:")
print([value * 2 for value in python_list])

# Press shift+enter to run the code
```

# 3. Why NumPy Is Used

## 1) Vectorization

Data science often requires the same calculation to be applied to many values.

<strong>Vectorization</strong> means writing operations for whole arrays instead of manually looping through one value at a time.

Vectorized code is useful because:

- It is concise and readable for numerical transformations.
- It often resembles mathematical notation.
- It lets NumPy run optimized lower-level routines.
- It reduces repetitive indexing logic in Python code.

```python
# Vectorized calculation example
prices = np.array([100, 250, 80, 120, 300])
discount_rate = 0.10

sale_prices = prices * (1 - discount_rate)

print("Original prices:", prices)
print("Sale prices:", sale_prices)

# Press shift+enter to run the code
```

## 2) Broadcasting

<strong>Broadcasting</strong> is NumPy's rule system for operating on arrays with compatible but different shapes.

For example, a one-dimensional array of subject bonuses can be added to every row of a two-dimensional score table.

```python
# Broadcasting example
scores = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [95, 88, 92],
])
bonus_by_subject = np.array([5, 0, 10])

adjusted_scores = scores + bonus_by_subject

print("Scores:")
print(scores)
print()
print("Bonus by subject:", bonus_by_subject)
print()
print("Adjusted scores:")
print(adjusted_scores)

# Press shift+enter to run the code
```

## 3) Speed And Memory

NumPy operations can be faster for large numerical workloads because much of the work is performed by optimized compiled code.

NumPy arrays can also be more memory-efficient than Python lists because array values are stored in a compact typed block.

<div class="note"><strong>Important:</strong> Tiny examples may not show meaningful performance differences. Benchmark with realistic data sizes before making performance decisions.</div>

```python
# Simple timing comparison with a moderately sized example
from timeit import timeit

values_list = list(range(100_000))
values_array = np.arange(100_000)

list_time = timeit(lambda: [value * 2 for value in values_list], number=100)
array_time = timeit(lambda: values_array * 2, number=100)

print(f"List time: {list_time:.4f} seconds")
print(f"NumPy time: {array_time:.4f} seconds")
print("List object size is not directly comparable with NumPy array memory.")
print("NumPy array bytes:", values_array.nbytes)

# Press shift+enter to run the code
```

# 4. Core Array Ideas

## 1) Shape, Axis, And `dtype`

NumPy becomes easier when you pay attention to three ideas:

| Idea | Meaning | Example |
|---|---|---|
| Shape | The size of each dimension | `(3, 4)` means 3 rows and 4 columns |
| Axis | A direction of calculation | `axis=0` works down columns, `axis=1` works across rows |
| `dtype` | The stored data type | Integer, float, boolean, and others |

These ideas appear repeatedly in array creation, indexing, aggregation, broadcasting, and modeling workflows.

```python
# Inspect shape, axis behavior, and dtype
measurements = np.array([
    [1.5, 2.0, 3.5],
    [2.5, 3.0, 4.5],
    [3.5, 4.0, 5.5],
])

print("Measurements:")
print(measurements)
print("Shape:", measurements.shape)
print("Dimensions:", measurements.ndim)
print("Data type:", measurements.dtype)
print()
print("Column means with axis=0:", measurements.mean(axis=0))
print("Row means with axis=1:", measurements.mean(axis=1))

# Press shift+enter to run the code
```

# 5. Practical Judgment

## 1) When To Use NumPy

NumPy is a good choice when the work is mainly numerical and array-based.

| Situation | Why NumPy helps |
|---|---|
| Many numeric values need the same operation | Array operations avoid repetitive manual loops |
| Data has vector or matrix structure | NumPy represents vectors and matrices naturally |
| Performance matters for numerical code | Optimized operations can be much faster |
| Other libraries expect array input | NumPy arrays are widely supported |
| You need reproducible numeric experiments | NumPy provides random number generation and numerical utilities |

Common beginner examples include scores, prices, measurements, images, simulation results, and model features.

## 2) Limitations And Trade-Offs

NumPy is powerful, but it is not always the best tool for every task.

| Limitation | Practical meaning |
|---|---|
| Same-type arrays | Mixed text, dates, and categories may be easier in pandas or Polars |
| No row and column labels by default | Labeled table analysis is usually easier in dataframe libraries |
| Memory-bound for very large data | Data larger than memory may require chunking, databases, Spark, or other tools |
| Requires array thinking | Shape, axis, and broadcasting take practice |

A practical rule: use NumPy for numerical arrays, use dataframe libraries for labeled tables, and use specialized tools when the data is too large or too domain-specific for memory-based array workflows.

```python
# Practical helper: inspect any NumPy array quickly
def describe_array(array):
    """Print basic metadata for a NumPy array."""
    print("shape:", array.shape)
    print("dtype:", array.dtype)
    print("dimensions:", array.ndim)
    print("size:", array.size)

example = np.arange(12).reshape(3, 4)
describe_array(example)

# Press shift+enter to run the code
```

# 4. Exploration

Use this section to understand what NumPy is, why arrays are useful, and how NumPy supports common data science workflows.

## What Is NumPy?

NumPy is a Python library for numerical and scientific computing. The name comes from Numerical Python.

It provides:

- A multidimensional array object called an N-dimensional array, or `ndarray`
- Derived objects, such as masked arrays and matrices
- Fast routines for array operations, including mathematics, logic, shape manipulation, sorting, selection, input/output, Fourier transforms, linear algebra, statistics, and random simulation

NumPy is commonly used for:

- Storing numerical data in arrays
- Performing calculations across many values
- Representing vectors, matrices, and higher-dimensional data
- Preparing data for machine learning and scientific computing
- Providing a foundation for other data science libraries

More details are available in the official NumPy documentation: https://numpy.org/doc/stable/user/whatisnumpy.html

In beginner terms, NumPy helps Python work with many numbers efficiently.

## List vs. NumPy Array

Python lists can store many values, but they are general-purpose containers. They can hold mixed data types and are designed for flexibility.

NumPy arrays are designed for numerical work. They usually store one data type in a compact structure, which makes mathematical operations faster and more consistent.

| Container | Main purpose | Typical data | Strength |
|---|---|---|---|
| Python list | General collection | Mixed or flexible data | Easy and flexible |
| NumPy array | Numerical computing | Same-type numeric data | Fast array operations |

## Why NumPy Is Used

Data science often requires the same calculation to be applied to many values.

<strong>1. Vectorization</strong>  
NumPy encourages thinking in terms of whole arrays instead of one value at a time. This usually means less explicit looping and indexing.

- Vectorized code is more concise and easier to read
- Fewer lines of code generally means fewer bugs
- The code more closely resembles standard mathematical notation
- Vectorized NumPy code is often more Pythonic than repetitive loops

<strong>2. Broadcasting</strong>  
Broadcasting is the mechanism that allows NumPy to perform arithmetic operations on arrays with different shapes.

Refer to the official documentation for details: https://numpy.org/doc/stable/user/basics.broadcasting.html#basics-broadcasting



<strong>3. Speed</strong>  
Because of vectorization and broadcasting, NumPy operations can be much faster for large numerical workloads.

<strong>4. Memory Efficiency</strong>  
NumPy arrays store values in a compact format. For large numerical datasets, this can use less memory than Python lists.

<strong>5. Ecosystem Compatibility</strong>  
Many data science libraries either use NumPy internally or accept NumPy arrays as inputs.

NumPy provides a practical bridge between readable Python code and efficient lower-level numerical operations.

```python
# Concept example: apply the same operation to many values
list_result = [value * 2 for value in python_list]
array_result = numpy_array * 2

print("List result:", list_result)
print("Array result:", array_result)

print("-" * 20)
print("Execution time with list:")
%timeit [x * 2 for x in python_list]
print("Execution time with NumPy array:")
%timeit numpy_array * 2

print("Which was faster?")

# Press shift+enter to run the code
```

## Exploration Notes

| Finding | Evidence | Action |
|---|---|---|
| NumPy arrays are built for numerical data | Arrays have shape, dimension, and dtype metadata | Use arrays when data is numeric and array-based |
| Vectorization changes how code is written | `numpy_array * 2` applies the operation to all values | Prefer array expressions for numerical transformations |
| Small examples may not show speed advantages | Timing overhead can dominate tiny arrays | Benchmark with realistic data sizes before making performance claims |

# 5. Development

## NumPy In The Data Science Stack

NumPy is often described as a foundation library because many tools are built around its array model.

| Tool | Relationship with NumPy |
|---|---|
| pandas | Uses NumPy-like arrays for many tabular data operations |
| SciPy | Builds scientific computing tools on top of NumPy arrays |
| scikit-learn | Commonly accepts NumPy arrays for features and targets |
| Matplotlib | Plots numerical data stored in NumPy arrays or similar structures |
| Tensor libraries | Share similar array concepts such as shape, dtype, and vectorized operations |

Learning NumPy makes it easier to understand the tools that come after it.

## When To Use NumPy

NumPy is a good choice when the work is mainly numerical and array-based. Here are typical situations where NumPy helps.

| Situation | Why NumPy helps |
|---|---|
| Many numeric values need the same operation | Array operations avoid repetitive manual loops |
| Data has matrix or vector structure | NumPy represents vectors and matrices naturally |
| Performance matters for numerical code | Optimized operations can be much faster |
| Other libraries expect array input | NumPy arrays are widely supported |
| You need reproducible numeric experiments | NumPy provides tools for random number generation and numerical workflows |

Common beginner examples include arrays of scores, prices, measurements, images, and model features.

## Limitations And Trade-Offs

NumPy is powerful, but it is not always the best tool for every use case.

| Limitation | Practical meaning |
|---|---|
| Same-type arrays | Mixed text, dates, and categories may be easier in pandas |
| Lower-level than pandas | NumPy does not provide labeled rows and columns by default |
| Memory-bound for very large data | Data larger than memory may require chunking, databases, Spark, or other tools |
| Requires array thinking | Beginners may need time to get comfortable with shape, axis, and broadcasting concepts |

A practical rule: use NumPy for numerical arrays, use pandas for labeled tables, and use specialized tools when the data is too large or too domain-specific for memory-based array workflows.

```python
# Helper function for simple NumPy inspection
def describe_array(array):
    """Print basic metadata for a NumPy array."""
    print("shape:", array.shape)
    print("dtype:", array.dtype)
    print("dimensions:", array.ndim)

describe_array(numpy_array)

# Press shift+enter to run the code
```
